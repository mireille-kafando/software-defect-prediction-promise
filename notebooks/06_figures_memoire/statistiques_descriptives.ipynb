```python
import pandas as pd
import os
from pathlib import Path

# Liste des chemins de fichiers CSV pour l'entrainement
train_file_paths = [
    '/content/drive/MyDrive/bug/bug-data/jedit/jedit-3.2.csv',
    '/content/drive/MyDrive/bug/bug-data/jedit/jedit-4.0.csv',
    '/content/drive/MyDrive/bug/bug-data/jedit/jedit-4.1.csv',
    '/content/drive/MyDrive/bug/bug-data/jedit/jedit-4.2.csv',
    '/content/drive/MyDrive/bug/bug-data/jedit/jedit-4.3.csv',
    '/content/drive/MyDrive/bug/bug-data/ant/ant-1.3.csv',
    '/content/drive/MyDrive/bug/bug-data/ant/ant-1.4.csv',
    '/content/drive/MyDrive/bug/bug-data/ant/ant-1.5.csv',
    '/content/drive/MyDrive/bug/bug-data/ant/ant-1.6.csv',
    '/content/drive/MyDrive/bug/bug-data/ant/ant-1.7.csv',
    '/content/drive/MyDrive/bug/bug-data/camel/camel-1.0.csv',
    '/content/drive/MyDrive/bug/bug-data/camel/camel-1.2.csv',
    '/content/drive/MyDrive/bug/bug-data/camel/camel-1.4.csv',
    '/content/drive/MyDrive/bug/bug-data/camel/camel-1.6.csv',
    '/content/drive/MyDrive/bug/bug-data/ivy/ivy-1.0.csv',
    '/content/drive/MyDrive/bug/bug-data/ivy/ivy-1.1.csv',
    '/content/drive/MyDrive/bug/bug-data/ivy/ivy-1.2.csv',
    '/content/drive/MyDrive/bug/bug-data/log4j/log4j-1.0.csv',
    '/content/drive/MyDrive/bug/bug-data/log4j/log4j-1.1.csv',
    '/content/drive/MyDrive/bug/bug-data/log4j/log4j-1.2.csv',
    '/content/drive/MyDrive/bug/bug-data/lucene/lucene-2.0.csv',
    '/content/drive/MyDrive/bug/bug-data/lucene/lucene-2.2.csv',
    '/content/drive/MyDrive/bug/bug-data/lucene/lucene-2.4.csv',
    '/content/drive/MyDrive/bug/bug-data/poi/poi-1.5.csv',
    '/content/drive/MyDrive/bug/bug-data/poi/poi-2.0.csv',
    '/content/drive/MyDrive/bug/bug-data/poi/poi-2.5.csv',
    '/content/drive/MyDrive/bug/bug-data/poi/poi-3.0.csv',
    '/content/drive/MyDrive/bug/bug-data/synapse/synapse-1.0.csv',
    '/content/drive/MyDrive/bug/bug-data/synapse/synapse-1.1.csv',
    '/content/drive/MyDrive/bug/bug-data/synapse/synapse-1.2.csv',
    '/content/drive/MyDrive/bug/bug-data/velocity/velocity-1.4.csv',
    '/content/drive/MyDrive/bug/bug-data/velocity/velocity-1.5.csv',
    '/content/drive/MyDrive/bug/bug-data/velocity/velocity-1.6.csv',
    '/content/drive/MyDrive/bug/bug-data/xalan/xalan-2.4.csv',
    '/content/drive/MyDrive/bug/bug-data/xalan/xalan-2.5.csv',
    '/content/drive/MyDrive/bug/bug-data/xalan/xalan-2.6.csv',
    '/content/drive/MyDrive/bug/bug-data/xalan/xalan-2.7.csv',
    '/content/drive/MyDrive/bug/bug-data/xerces/xerces-1.1.csv',
    '/content/drive/MyDrive/bug/bug-data/xerces/xerces-1.2.csv',
    '/content/drive/MyDrive/bug/bug-data/xerces/xerces-1.3.csv',
    '/content/drive/MyDrive/bug/bug-data/xerces/xerces-1.4.4.csv',
]

def analyze_individual_files():
    """Analyse chaque fichier individuellement"""
    print("="*80)
    print("ANALYSE INDIVIDUELLE DES FICHIERS")
    print("="*80)

    for file_path in train_file_paths:
        try:
            # Extraire le nom du fichier
            file_name = Path(file_path).name

            # Charger le fichier
            df = pd.read_csv(file_path)

            print(f"\n📁 Fichier: {file_name}")
            print(f"   Dimensions: {df.shape[0]} lignes × {df.shape[1]} colonnes")

            # Informations générales
            print(f"   Colonnes: {list(df.columns)}")

            # Vérifier s'il y a une colonne 'bug' ou similaire pour les classes
            bug_columns = [col for col in df.columns if 'bug' in col.lower() or 'defect' in col.lower() or 'class' in col.lower()]
            if bug_columns:
                for bug_col in bug_columns:
                    if df[bug_col].dtype in ['int64', 'float64', 'bool'] or df[bug_col].nunique() <= 10:
                        value_counts = df[bug_col].value_counts()
                        print(f"   Distribution {bug_col}: {dict(value_counts)}")

            # Statistiques pour les colonnes numériques
            numeric_cols = df.select_dtypes(include=['number']).columns
            if len(numeric_cols) > 0:
                print(f"   Colonnes numériques: {len(numeric_cols)}")
                print(f"   Valeurs manquantes: {df.isnull().sum().sum()}")

                # Statistiques descriptives avec describe()
                print(f"\n   📊 Statistiques descriptives:")
                desc_stats = df[numeric_cols].describe()
                print(desc_stats.round(3))

        except Exception as e:
            print(f"❌ Erreur lors du chargement de {file_path}: {str(e)}")

def analyze_combined_dataset():
    """Analyse du dataset combiné"""
    print("\n" + "="*80)
    print("ANALYSE DU DATASET COMBINÉ")
    print("="*80)

    all_dataframes = []
    file_info = []

    for file_path in train_file_paths:
        try:
            df = pd.read_csv(file_path)
            file_name = Path(file_path).name

            # Ajouter une colonne pour identifier la source
            df['source_file'] = file_name
            df['project'] = file_name.split('-')[0]  # Extraire le nom du projet

            all_dataframes.append(df)
            file_info.append({
                'file': file_name,
                'rows': df.shape[0],
                'cols': df.shape[1],
                'project': file_name.split('-')[0]
            })

        except Exception as e:
            print(f"❌ Erreur lors du chargement de {file_path}: {str(e)}")

    if all_dataframes:
        # Combiner tous les DataFrames
        combined_df = pd.concat(all_dataframes, ignore_index=True, sort=False)

        print(f"\n📊 STATISTIQUES GLOBALES:")
        print(f"   Nombre total de fichiers: {len(file_info)}")
        print(f"   Dimensions combinées: {combined_df.shape[0]} lignes × {combined_df.shape[1]} colonnes")
        print(f"   Projets inclus: {combined_df['project'].nunique()}")
        print(f"   Liste des projets: {sorted(combined_df['project'].unique())}")

        # Statistiques par projet
        print(f"\n📈 RÉPARTITION PAR PROJET:")
        project_stats = combined_df.groupby('project').agg({
            'source_file': 'count',
            combined_df.columns[0]: 'count'  # Utiliser la première colonne pour compter les lignes
        }).rename(columns={'source_file': 'nb_fichiers', combined_df.columns[0]: 'nb_lignes'})

        for project, stats in project_stats.iterrows():
            print(f"   {project}: {stats['nb_fichiers']} fichiers, {stats['nb_lignes']} lignes")

        # Colonnes communes
        print(f"\n📋 ANALYSE DES COLONNES:")
        all_columns = set()
        for df in all_dataframes:
            all_columns.update(df.columns)
        all_columns.discard('source_file')
        all_columns.discard('project')

        # Trouver les colonnes présentes dans tous les fichiers
        common_columns = set(all_dataframes[0].columns)
        for df in all_dataframes[1:]:
            common_columns = common_columns.intersection(set(df.columns))
        common_columns.discard('source_file')
        common_columns.discard('project')

        print(f"   Colonnes uniques totales: {len(all_columns)}")
        print(f"   Colonnes communes à tous les fichiers: {len(common_columns)}")
        print(f"   Colonnes communes: {sorted(list(common_columns))}")

        # Statistiques descriptives sur les colonnes numériques communes
        if common_columns:
            numeric_common = combined_df[list(common_columns)].select_dtypes(include=['number']).columns
            if len(numeric_common) > 0:
                print(f"\n📊 STATISTIQUES DESCRIPTIVES (colonnes numériques communes):")
                desc_stats = combined_df[numeric_common].describe()
                print(desc_stats.round(3))

                # Valeurs manquantes
                print(f"\n🔍 VALEURS MANQUANTES:")
                missing_stats = combined_df[numeric_common].isnull().sum()
                missing_percent = (missing_stats / len(combined_df) * 100).round(2)
                for col in numeric_common:
                    print(f"   {col}: {missing_stats[col]} ({missing_percent[col]}%)")

        return combined_df
    else:
        print("❌ Aucun fichier n'a pu être chargé.")
        return None

def create_summary_report():
    """Crée un rapport de synthèse"""
    print("\n" + "="*80)
    print("RAPPORT DE SYNTHÈSE")
    print("="*80)

    successful_files = 0
    total_rows = 0
    projects = set()

    for file_path in train_file_paths:
        try:
            df = pd.read_csv(file_path)
            successful_files += 1
            total_rows += len(df)
            project_name = Path(file_path).name.split('-')[0]
            projects.add(project_name)
        except:
            continue

    print(f"✅ Fichiers chargés avec succès: {successful_files}/{len(train_file_paths)}")
    print(f"📊 Total des lignes de données: {total_rows:,}")
    print(f"🏗️  Nombre de projets: {len(projects)}")
    print(f"📂 Projets: {', '.join(sorted(projects))}")
    print(f"📈 Moyenne de lignes par fichier: {total_rows//successful_files if successful_files > 0 else 0:,}")

def get_detailed_describe(file_path):
    """Fonction pour obtenir des statistiques describe() détaillées d'un fichier spécifique"""
    try:
        df = pd.read_csv(file_path)
        file_name = Path(file_path).name

        print(f"\n{'='*60}")
        print(f"DESCRIBE DÉTAILLÉ - {file_name}")
        print(f"{'='*60}")

        # Informations générales
        print(f"Dimensions: {df.shape}")
        print(f"Colonnes: {list(df.columns)}")

        # Describe pour toutes les colonnes (numériques et catégorielles)
        print(f"\n📊 DESCRIBE COMPLET:")
        print(df.describe(include='all').round(3))

        # Describe uniquement pour les colonnes numériques (plus détaillé)
        numeric_cols = df.select_dtypes(include=['number']).columns
        if len(numeric_cols) > 0:
            print(f"\n🔢 DESCRIBE NUMÉRIQUES SEULEMENT:")
            print(df[numeric_cols].describe().round(3))

            # Percentiles supplémentaires
            print(f"\n📈 PERCENTILES SUPPLÉMENTAIRES:")
            percentiles = [0.01, 0.05, 0.10, 0.90, 0.95, 0.99]
            extra_stats = df[numeric_cols].quantile(percentiles)
            print(extra_stats.round(3))

        # Informations sur les valeurs manquantes
        print(f"\n❓ VALEURS MANQUANTES:")
        missing_info = pd.DataFrame({
            'Count': df.isnull().sum(),
            'Percentage': (df.isnull().sum() / len(df) * 100).round(2)
        })
        print(missing_info[missing_info['Count'] > 0])

        # Types de données
        print(f"\n📋 TYPES DE DONNÉES:")
        print(df.dtypes)

        return df

    except Exception as e:
        print(f"❌ Erreur lors du chargement de {file_path}: {str(e)}")
        return None

def analyze_all_describes():
    """Affiche le describe() pour tous les fichiers"""
    print("="*80)
    print("DESCRIBE() POUR TOUS LES FICHIERS")
    print("="*80)

    for i, file_path in enumerate(train_file_paths, 1):
        print(f"\n[{i}/{len(train_file_paths)}] Processing: {Path(file_path).name}")
        get_detailed_describe(file_path)

        # Pause entre les fichiers pour la lisibilité
        if i < len(train_file_paths):
            print(f"\n{'-'*40}")

def quick_describe_summary():
    """Résumé rapide des describe() de tous les fichiers"""
    print("="*80)
    print("RÉSUMÉ RAPIDE - DESCRIBE() TOUS LES FICHIERS")
    print("="*80)

    summary_stats = []

    for file_path in train_file_paths:
        try:
            df = pd.read_csv(file_path)
            file_name = Path(file_path).name

            numeric_cols = df.select_dtypes(include=['number']).columns
            if len(numeric_cols) > 0:
                desc = df[numeric_cols].describe()

                summary_stats.append({
                    'File': file_name,
                    'Rows': df.shape[0],
                    'Numeric_Cols': len(numeric_cols),
                    'Mean_of_Means': desc.loc['mean'].mean(),
                    'Mean_of_Stds': desc.loc['std'].mean(),
                    'Total_Missing': df.isnull().sum().sum()
                })
        except Exception as e:
            print(f"❌ Erreur avec {file_path}: {str(e)}")

    if summary_stats:
        summary_df = pd.DataFrame(summary_stats)
        print(summary_df.round(3))

        print(f"\n📊 STATISTIQUES GLOBALES DU RÉSUMÉ:")
        print(f"Total fichiers analysés: {len(summary_stats)}")
        print(f"Total lignes: {summary_df['Rows'].sum():,}")
        print(f"Moyenne colonnes numériques par fichier: {summary_df['Numeric_Cols'].mean():.1f}")
        print(f"Total valeurs manquantes: {summary_df['Total_Missing'].sum():,}")

# Exécution des analyses
if __name__ == "__main__":
    print("🔍 ANALYSE DES DONNÉES DE BUGS - DÉBUT")
    print("="*80)

    # Option 1: Analyse rapide avec describe() intégré
    analyze_individual_files()

    # Option 2: Résumé rapide des describe()
    quick_describe_summary()

    # Option 3: Describe détaillé
    #analyze_all_describes()

    # Analyse combinée et rapport de synthèse
    combined_data = analyze_combined_dataset()
    create_summary_report()

    print(f"\n✅ ANALYSE TERMINÉE")
    print("="*80)

```
