````python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from typing import List, Dict

class BugMetricsAnalyzer:
    """
    Classe pour analyser les métriques de bugs à travers différentes versions de logiciel.
    """
    def __init__(self, train_paths: List[str], val_path: str, versions: List[str]):
        """
        Initialise l'analyseur avec les chemins des fichiers et les versions.

        Args:
            train_paths: Liste des chemins vers les fichiers d'entraînement
            val_path: Chemin vers le fichier de validation
            versions: Liste des versions à analyser
        """
        self.train_paths = train_paths
        self.val_path = val_path
        self.versions = versions
        self.metrics = ["loc", "noc", "cbo", "max_cc"]
        self.df_all = None

    def load_data(self) -> None:
        """Charge et prépare les données pour l'analyse."""
        # Chargement des données d'entraînement
        train_dfs = []
        for file_path, version in zip(self.train_paths, self.versions[:-1]):
            df = pd.read_csv(file_path, usecols=lambda x: x != 'name')
            df['version'] = version
            train_dfs.append(df)

        # Chargement des données de validation
        df_val = pd.read_csv(self.val_path, usecols=lambda x: x != 'name')
        df_val['version'] = self.versions[-1]

        # Concaténation de toutes les données
        self.df_all = pd.concat(train_dfs + [df_val], ignore_index=True)

    def calculate_statistics(self, metric: str) -> pd.DataFrame:
        """
        Calcule les statistiques descriptives pour une métrique donnée.

        Args:
            metric: Nom de la métrique à analyser

        Returns:
            DataFrame contenant les statistiques par version
        """
        stats_list = []
        for version in self.versions:
            version_data = self.df_all[self.df_all['version'] == version][metric]
            stats = {
                'Version': version,
                'Moyenne': np.mean(version_data),
                'Médiane': np.median(version_data),
                'Min': np.min(version_data),
                'Max': np.max(version_data),
                'Écart-type': np.std(version_data)
            }
            stats_list.append(stats)
        return pd.DataFrame(stats_list).round(2)

    def plot_histograms(self, metric: str) -> None:
        """
        Génère les histogrammes pour une métrique donnée.

        Args:
            metric: Nom de la métrique à visualiser
        """
        fig, axs = plt.subplots(1, len(self.versions), figsize=(8, 4), sharey=True)
        fig.suptitle(f"Histogramme de {metric} par version")

        for i, version in enumerate(self.versions):
            version_data = self.df_all[self.df_all['version'] == version][metric]
            axs[i].hist(version_data, bins=20, color='skyblue', edgecolor='black')
            axs[i].set_title(f'Version {version}')
            axs[i].set_xlabel(metric)
            axs[i].set_ylabel("Fréquence")

        plt.tight_layout(rect=[0, 0, 1, 0.96])
        plt.show()

    def plot_evolution(self, metric: str) -> None:
        """
        Génère la courbe d'évolution pour une métrique donnée.

        Args:
            metric: Nom de la métrique à visualiser
        """
        plt.figure(figsize=(4, 2))
        mean_values = self.df_all.groupby('version')[metric].mean()
        plt.plot(mean_values.index, mean_values.values, marker='o', color='b')
        plt.xlabel("Version")
        plt.ylabel(f"Moyenne de {metric}")
        plt.title(f"Évolution de la moyenne de {metric} par version")
        plt.grid()
        plt.show()

    def analyze_metrics(self) -> None:
        """Lance l'analyse complète pour toutes les métriques."""
        self.load_data()

        for metric in self.metrics:
            # Génération des visualisations
            self.plot_histograms(metric)
            self.plot_evolution(metric)

            # Calcul et affichage des statistiques
            print(f"\nStatistiques pour {metric.upper()}:")
            stats_df = self.calculate_statistics(metric)
            print(stats_df)
            print("\n" + "=" * 60 + "\n")

def main():
    """Fonction principale pour exécuter l'analyse."""
    # Configuration des chemins et versions
import pandas as pd

# Liste des chemins de fichiers CSV pour l'entrainement
train_file_paths = [
    '/content/drive/MyDrive/bug/bug-data/jedit/jedit-3.2.csv',
    '/content/drive/MyDrive/bug/bug-data/jedit/jedit-4.0.csv',
    '/content/drive/MyDrive/bug/bug-data/jedit/jedit-4.1.csv',
    '/content/drive/MyDrive/bug/bug-data/jedit/jedit-4.2.csv',
    '/content/drive/MyDrive/bug/bug-data/jedit/jedit-4.3.csv',
    '/content/drive/MyDrive/bug/bug-data/ant/ant-1.3.csv',
    '/content/drive/MyDrive/bug/bug-data/ant/ant-1.4.csv',
    '/content/drive/MyDrive/bug/bug-data/ant/ant-1.5.csv',
    '/content/drive/MyDrive/bug/bug-data/ant/ant-1.6.csv',
    '/content/drive/MyDrive/bug/bug-data/ant/ant-1.7.csv',
    '/content/drive/MyDrive/bug/bug-data/camel/camel-1.0.csv',
    '/content/drive/MyDrive/bug/bug-data/camel/camel-1.2.csv',
    '/content/drive/MyDrive/bug/bug-data/camel/camel-1.4.csv',
    '/content/drive/MyDrive/bug/bug-data/camel/camel-1.6.csv',
    '/content/drive/MyDrive/bug/bug-data/ivy/ivy-1.0.csv',
    '/content/drive/MyDrive/bug/bug-data/ivy/ivy-1.1.csv',
    '/content/drive/MyDrive/bug/bug-data/ivy/ivy-1.2.csv',
    '/content/drive/MyDrive/bug/bug-data/log4j/log4j-1.0.csv',
    '/content/drive/MyDrive/bug/bug-data/log4j/log4j-1.1.csv',
    '/content/drive/MyDrive/bug/bug-data/log4j/log4j-1.2.csv',
    '/content/drive/MyDrive/bug/bug-data/lucene/lucene-2.0.csv',
    '/content/drive/MyDrive/bug/bug-data/lucene/lucene-2.2.csv',
    '/content/drive/MyDrive/bug/bug-data/lucene/lucene-2.4.csv',
    '/content/drive/MyDrive/bug/bug-data/poi/poi-1.5.csv',
    '/content/drive/MyDrive/bug/bug-data/poi/poi-2.0.csv',
    '/content/drive/MyDrive/bug/bug-data/poi/poi-2.5.csv',
    '/content/drive/MyDrive/bug/bug-data/poi/poi-3.0.csv',
    '/content/drive/MyDrive/bug/bug-data/synapse/synapse-1.0.csv',
    '/content/drive/MyDrive/bug/bug-data/synapse/synapse-1.1.csv',
    '/content/drive/MyDrive/bug/bug-data/synapse/synapse-1.2.csv',
    '/content/drive/MyDrive/bug/bug-data/velocity/velocity-1.4.csv',
    '/content/drive/MyDrive/bug/bug-data/velocity/velocity-1.5.csv',
    '/content/drive/MyDrive/bug/bug-data/velocity/velocity-1.6.csv',
    '/content/drive/MyDrive/bug/bug-data/xalan/xalan-2.4.csv',
    '/content/drive/MyDrive/bug/bug-data/xalan/xalan-2.5.csv',
    '/content/drive/MyDrive/bug/bug-data/xalan/xalan-2.6.csv',
    '/content/drive/MyDrive/bug/bug-data/xerces/xerces-1.1.csv',
    '/content/drive/MyDrive/bug/bug-data/xerces/xerces-1.2.csv',
    '/content/drive/MyDrive/bug/bug-data/xerces/xerces-1.3.csv',
    '/content/drive/MyDrive/bug/bug-data/xerces/xerces-1.4.4.csv',

]

    versions = ['1.1', '1.2', '1.3', '1.4.4']

    # Création et exécution de l'analyseur
    analyzer = BugMetricsAnalyzer(train_file_paths, versions)
    analyzer.analyze_metrics()

if __name__ == "__main__":
    main()
````

````python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from typing import List, Dict

class BugMetricsAnalyzer:
    """
    Classe pour analyser les métriques de bugs à travers différentes versions de logiciel.
    """
    def __init__(self, train_paths: List[str], val_path: str, versions: List[str]):
        """
        Initialise l'analyseur avec les chemins des fichiers et les versions.

        Args:
            train_paths: Liste des chemins vers les fichiers d'entraînement
            val_path: Chemin vers le fichier de validation
            versions: Liste des versions à analyser
        """
        self.train_paths = train_paths
        self.val_path = val_path
        self.versions = versions
        self.metrics = ["loc", "noc", "cbo", "max_cc"]
        self.df_all = None

    def load_data(self) -> None:
        """Charge et prépare les données pour l'analyse."""
        # Chargement des données d'entraînement
        train_dfs = []
        for file_path, version in zip(self.train_paths, self.versions[:-1]):
            df = pd.read_csv(file_path, usecols=lambda x: x != 'name')
            df['version'] = version
            train_dfs.append(df)

        # Chargement des données de validation
        df_val = pd.read_csv(self.val_path, usecols=lambda x: x != 'name')
        df_val['version'] = self.versions[-1]

        # Concaténation de toutes les données
        self.df_all = pd.concat(train_dfs + [df_val], ignore_index=True)

    def calculate_statistics(self, metric: str) -> pd.DataFrame:
        """
        Calcule les statistiques descriptives pour une métrique donnée.

        Args:
            metric: Nom de la métrique à analyser

        Returns:
            DataFrame contenant les statistiques par version
        """
        stats_list = []
        for version in self.versions:
            version_data = self.df_all[self.df_all['version'] == version][metric]
            stats = {
                'Version': version,
                'Moyenne': np.mean(version_data),
                'Médiane': np.median(version_data),
                'Min': np.min(version_data),
                'Max': np.max(version_data),
                'Écart-type': np.std(version_data)
            }
            stats_list.append(stats)
        return pd.DataFrame(stats_list).round(2)

    def plot_histograms(self, metric: str) -> None:
        """
        Génère les histogrammes pour une métrique donnée.

        Args:
            metric: Nom de la métrique à visualiser
        """
        fig, axs = plt.subplots(1, len(self.versions), figsize=(8, 4), sharey=True)
        fig.suptitle(f"Histogramme de {metric} par version")

        for i, version in enumerate(self.versions):
            version_data = self.df_all[self.df_all['version'] == version][metric]
            axs[i].hist(version_data, bins=20, color='skyblue', edgecolor='black')
            axs[i].set_title(f'Version {version}')
            axs[i].set_xlabel(metric)
            axs[i].set_ylabel("Fréquence")

        plt.tight_layout(rect=[0, 0, 1, 0.96])
        plt.show()

    def plot_evolution(self, metric: str) -> None:
        """
        Génère la courbe d'évolution pour une métrique donnée.

        Args:
            metric: Nom de la métrique à visualiser
        """
        plt.figure(figsize=(4, 2))
        mean_values = self.df_all.groupby('version')[metric].mean()
        plt.plot(mean_values.index, mean_values.values, marker='o', color='b')
        plt.xlabel("Version")
        plt.ylabel(f"Moyenne de {metric}")
        plt.title(f"Évolution de la moyenne de {metric} par version")
        plt.grid()
        plt.show()

    def analyze_metrics(self) -> None:
        """Lance l'analyse complète pour toutes les métriques."""
        self.load_data()

        for metric in self.metrics:
            # Génération des visualisations
            self.plot_histograms(metric)
            self.plot_evolution(metric)

            # Calcul et affichage des statistiques
            print(f"\nStatistiques pour {metric.upper()}:")
            stats_df = self.calculate_statistics(metric)
            print(stats_df)
            print("\n" + "=" * 60 + "\n")

def main():
    """Fonction principale pour exécuter l'analyse."""
    # Configuration des chemins et versions
    train_file_paths = [
    '/content/drive/MyDrive/bug/bug-data/xerces/xerces-1.1.csv',
    '/content/drive/MyDrive/bug/bug-data/xerces/xerces-1.2.csv',
    '/content/drive/MyDrive/bug/bug-data/xerces/xerces-1.3.csv',

    ]
    val_file_path = '/content/drive/MyDrive/bug/bug-data/xerces/xerces-1.4.4.csv'

    versions = ['1.1', '1.2', '1.3', '1.4.4']

    # Création et exécution de l'analyseur
    analyzer = BugMetricsAnalyzer(train_file_paths, val_file_path, versions)
    analyzer.analyze_metrics()

if __name__ == "__main__":
    main()
````

````python
import numpy as np
import pandas as pd
from scipy import stats
import matplotlib.pyplot as plt
from statsmodels.stats.multicomp import pairwise_tukeyhsd

# Données
data = {
    'Systèmes': ['Ant', 'Camel', 'Ivy', 'Jedit', 'Log4j', 'Lucène', 'Poi', 'Synapse', 'Velocity'],
    'Gmean_1er_cas': [0.7117, 0.5870, 0.7476, 0.6247, 0.5528, 0.5935, 0.7338, 0.7098, 0.6173],
    'roc_auc_1er_cas': [0.7974, 0.6358, 0.8129, 0.6305, 0.5999, 0.6685, 0.7880, 0.7275, 0.6725],
    'mean_acc_1er_cas': [0.654, 0.6495, 0.6525, 0.6580, 0.6593, 0.6571, 0.6490, 0.6530, 0.6579],

    'Gmean_2ème_cas': [0.6943, 0.5911, 0.6794, 0.6687, 0.4950, 0.5889, 0.7306, 0.5895, 0.5112],
    'roc_auc_2ème_cas': [0.7408, 0.6474, 0.7478, 0.6766, 0.5443, 0.6484, 0.7536, 0.7141, 0.5728],
    'mean_acc_2ème_cas': [0.6809, 0.6891, 0.6878, 0.6710, 0.6805, 0.6745, 0.6733, 0.6763, 0.6864],

    'Gmean_3ème_cas': [0.6825, 0.5853, 0.6585, 0.6182, 0.4993, 0.6023, 0.6776, 0.6246, 0.5204],
    'roc_auc_3ème_cas': [0.7493, 0.6569, 0.7517, 0.6526, 0.5188, 0.6534, 0.7628, 0.7165, 0.5282],
    'mean_acc_3ème_cas': [0.6986, 0.6764, 0.6908, 0.6864, 0.6928, 0.6691, 0.6870, 0.6845, 0.6828]
}

df = pd.DataFrame(data)

def test_normalite(data, nom):
    statistic, p_value = stats.shapiro(data)
    print(f"Test de normalité pour {nom}:")
    print(f"Statistique: {statistic:.4f}")
    print(f"p-value: {p_value:.4f}")
    print(f"Distribution {'normale' if p_value > 0.05 else 'non normale'}\n")
    return p_value > 0.05

def anova_test(data1, data2, data3, nom):
    f_statistic, p_value = stats.f_oneway(data1, data2, data3)
    print(f"Résultats ANOVA pour {nom}:")
    print(f"F-statistic: {f_statistic:.4f}")
    print(f"p-value: {p_value:.4f}")
    print(f"Différence {'significative' if p_value < 0.05 else 'non significative'}\n")
    return p_value < 0.05

def kruskal_wallis_test(data1, data2, data3, nom):
    h_statistic, p_value = stats.kruskal(data1, data2, data3)
    print(f"Résultats Kruskal-Wallis pour {nom}:")
    print(f"H-statistic: {h_statistic:.4f}")
    print(f"p-value: {p_value:.4f}")
    print(f"Différence {'significative' if p_value < 0.05 else 'non significative'}\n")
    return p_value < 0.05

def tukey_hsd(data, group_labels, nom):
    tukey_results = pairwise_tukeyhsd(data, group_labels)
    print(f"Résultats Tukey HSD pour {nom}:")
    print(tukey_results)
    print("\n")

def analyse_data(data_cols, prefix):
    data = [df[col] for col in data_cols]
    normal = all([test_normalite(d, col) for d, col in zip(data, data_cols)])

    if normal:
        significant = anova_test(*data, prefix)
        if significant:
            all_data = np.concatenate(data)
            group_labels = np.repeat([f'{i+1}' for i in range(len(data_cols))], len(data[0]))
            tukey_hsd(all_data, group_labels, prefix)
    else:
        significant = kruskal_wallis_test(*data, prefix)

    # Visualisation
    fig, axes = plt.subplots(1, 3, figsize=(6, 3))
    fig.suptitle(f"Distribution des valeurs {prefix}")

    for i, (col, d) in enumerate(zip(data_cols, data)):
        axes[i].hist(d, bins=10)
        axes[i].set_title(f"{col}")
        axes[i].set_xlabel("Valeur")
        axes[i].set_ylabel("Fréquence")

    plt.tight_layout()
    plt.show()

# Analyse pour Gmean
gmean_cols = ['Gmean_1er_cas', 'Gmean_2ème_cas', 'Gmean_3ème_cas']
analyse_data(gmean_cols, "Gmean")

# Analyse pour roc_auc
roc_auc_cols = ['roc_auc_1er_cas', 'roc_auc_2ème_cas', 'roc_auc_3ème_cas']
analyse_data(roc_auc_cols, "roc_auc")

# Analyse pour mean_acc
mean_acc_cols = ['mean_acc_1er_cas', 'mean_acc_2ème_cas', 'mean_acc_3ème_cas']
analyse_data(mean_acc_cols, "mean_acc")
````

````python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from typing import List, Dict, Tuple
from collections import defaultdict
from scipy import stats

class BugMetricsAnalyzer:
    def __init__(self, train_paths: List[str]):
        self.train_paths = train_paths
        self.metrics = ["loc", "noc", "cbo", "max_cc"]
        self.metric_names = {
            "loc": "Lines of Code",
            "noc": "Number of Classes",
            "cbo": "Coupling Between Objects",
            "max_cc": "Maximum Cyclomatic Complexity"
        }
        self.df_all = None
        self.systems = self._extract_systems()

    def _extract_systems(self) -> Dict[str, List[str]]:
        """Extrait les systèmes et leurs versions à partir des chemins de fichiers."""
        systems_dict = defaultdict(list)
        for path in self.train_paths:
            parts = path.split('/')
            for part in parts:
                if '.csv' in part:
                    system = part.split('-')[0]
                    version = part.split('-')[1].replace('.csv', '')
                    systems_dict[system].append((version, path))

        for system in systems_dict:
            systems_dict[system].sort(key=lambda x: x[0])

        return dict(systems_dict)

    def load_data(self) -> None:
        """Charge et prépare les données pour l'analyse."""
        all_dfs = []
        for system, version_paths in self.systems.items():
            for version, path in version_paths:
                df = pd.read_csv(path, usecols=lambda x: x != 'name')
                df['version'] = version
                df['system'] = system
                all_dfs.append(df)

        self.df_all = pd.concat(all_dfs, ignore_index=True)

    def calculate_statistics(self, data: pd.Series) -> Dict[str, float]:
        """
        Calcule des statistiques détaillées pour une série de données.

        Args:
            data: Série de données à analyser

        Returns:
            Dict contenant les différentes statistiques calculées
        """
        stats_dict = {
            'variance': data.var(),
            'std_dev': data.std(),
            'mean': data.mean(),
            'median': data.median(),
            'cv': (data.std() / data.mean()) * 100 if data.mean() != 0 else np.nan,
            'q1': data.quantile(0.25),
            'q3': data.quantile(0.75),
            'iqr': data.quantile(0.75) - data.quantile(0.25),
            'skewness': stats.skew(data),
            'kurtosis': stats.kurtosis(data)
        }
        return stats_dict

    def perform_trend_analysis(self, values: np.ndarray) -> Dict[str, float]:
        """
        Effectue une analyse de tendance sur une série de valeurs.

        Args:
            values: Série de valeurs à analyser

        Returns:
            Dict contenant les résultats de l'analyse de tendance
        """
        x = np.arange(len(values))
        slope, intercept, r_value, p_value, std_err = stats.linregress(x, values)

        return {
            'slope': slope,
            'r_squared': r_value ** 2,
            'p_value': p_value,
            'trend': 'increasing' if slope > 0 else 'decreasing' if slope < 0 else 'stable'
        }

    def plot_metric_evolution_by_system(self, metric: str) -> None:
        """
        Génère un graphique amélioré montrant l'évolution de la variance et autres statistiques
        d'une métrique pour tous les systèmes.
        """
        plt.rcParams['figure.facecolor'] = 'white'
        plt.rcParams['axes.facecolor'] = 'white'
        plt.rcParams['axes.grid'] = True
        plt.rcParams['grid.color'] = '#E6E6E6'
        plt.rcParams['grid.linestyle'] = '--'

        fig = plt.figure(figsize=(20, 12))
        gs = plt.GridSpec(2, 2, height_ratios=[3, 1])
        ax_main = fig.add_subplot(gs[0, :])
        ax_stats = fig.add_subplot(gs[1, :])

        ax_main.grid(True, linestyle='--', alpha=0.7, zorder=0)
        ax_main.set_axisbelow(True)

        colors = plt.cm.tab20(np.linspace(0, 1, len(self.systems)))
        all_stats = []
        system_trends = {}

        for (system, version_paths), color in zip(self.systems.items(), colors):
            system_data = self.df_all[self.df_all['system'] == system]
            grouped_data = system_data.groupby('version')[metric]

            # Calcul des statistiques par version
            version_stats = []
            x_values = range(len(grouped_data))

            for _, group in grouped_data:
                stats_dict = self.calculate_statistics(group)
                version_stats.append(stats_dict)

            # Extraction des variances pour le graphique principal
            variance_values = [stats['variance'] for stats in version_stats]

            # Tracé de la ligne principale
            line = ax_main.plot(x_values, variance_values,
                              marker='o', label=system, color=color,
                              linewidth=2.5, markersize=8, zorder=2)

            # Ajout de l'intervalle de confiance
            std_dev_values = [stats['std_dev'] for stats in version_stats]
            ax_main.fill_between(x_values,
                               np.array(variance_values) - np.array(std_dev_values),
                               np.array(variance_values) + np.array(std_dev_values),
                               alpha=0.1, color=color, zorder=1)

            # Analyse de tendance
            trend_analysis = self.perform_trend_analysis(np.array(variance_values))
            system_trends[system] = trend_analysis

            # Stockage des statistiques pour affichage
            all_stats.extend(version_stats)

        # Configuration du graphique principal
        ax_main.set_title(f'Evolution of {self.metric_names[metric]} Variance\nAcross Different Versions',
                         fontsize=14, fontweight='bold', pad=20)
        ax_main.set_xlabel("Version Index", fontsize=12, labelpad=10)
        ax_main.set_ylabel(f"Variance of {self.metric_names[metric]}", fontsize=12, labelpad=10)

        # Tableau des statistiques
        stats_df = pd.DataFrame(all_stats)
        stats_summary = stats_df.agg(['mean', 'min', 'max', 'std']).round(2)

        # Création d'une table de statistiques
        cell_text = []
        for idx, row in stats_summary.iterrows():
            cell_text.append([f"{val:.2f}" for val in row])

        table = ax_stats.table(cellText=cell_text,
                             rowLabels=['Mean', 'Min', 'Max', 'Std'],
                             colLabels=stats_df.columns,
                             loc='center',
                             cellLoc='center')

        table.auto_set_font_size(False)
        table.set_fontsize(9)
        table.scale(1.2, 1.5)
        ax_stats.axis('off')

        # Ajout des informations de tendance
        trend_text = "Trend Analysis:\n"
        for system, trend in system_trends.items():
            trend_text += (f"{system}: {trend['trend']} (slope: {trend['slope']:.2f}, "
                         f"R²: {trend['r_squared']:.2f}, p-value: {trend['p_value']:.3f})\n")

        plt.figtext(1.02, 0.5, trend_text,
                   bbox=dict(facecolor='white', edgecolor='gray', alpha=0.8),
                   fontsize=10,
                   transform=ax_main.transAxes)

        # Légende
        legend = ax_main.legend(bbox_to_anchor=(1.05, 1),
                              loc='upper left',
                              title="Systems",
                              title_fontsize=12,
                              fontsize=10,
                              borderaxespad=0.)

        plt.tight_layout()
        plt.show()

    def analyze_all_metrics(self) -> None:
        """Lance l'analyse complète pour toutes les métriques."""
        self.load_data()

        for metric in self.metrics:
            self.plot_metric_evolution_by_system(metric)
            print(f"\nAnalyse terminée pour la métrique : {metric}")
            print("=" * 60)

def main():
    # Liste des chemins de fichiers CSV (inchangée)
    train_file_paths = [
        '/content/drive/MyDrive/bug/bug-data/jedit/jedit-3.2.csv',
        '/content/drive/MyDrive/bug/bug-data/jedit/jedit-4.0.csv',
        '/content/drive/MyDrive/bug/bug-data/jedit/jedit-4.1.csv',
        '/content/drive/MyDrive/bug/bug-data/jedit/jedit-4.2.csv',
        '/content/drive/MyDrive/bug/bug-data/jedit/jedit-4.3.csv',
        '/content/drive/MyDrive/bug/bug-data/ant/ant-1.3.csv',
        '/content/drive/MyDrive/bug/bug-data/ant/ant-1.4.csv',
        '/content/drive/MyDrive/bug/bug-data/ant/ant-1.5.csv',
        '/content/drive/MyDrive/bug/bug-data/ant/ant-1.6.csv',
        '/content/drive/MyDrive/bug/bug-data/ant/ant-1.7.csv',
        '/content/drive/MyDrive/bug/bug-data/camel/camel-1.0.csv',
        '/content/drive/MyDrive/bug/bug-data/camel/camel-1.2.csv',
        '/content/drive/MyDrive/bug/bug-data/camel/camel-1.4.csv',
        '/content/drive/MyDrive/bug/bug-data/camel/camel-1.6.csv',
        '/content/drive/MyDrive/bug/bug-data/ivy/ivy-1.0.csv',
        '/content/drive/MyDrive/bug/bug-data/ivy/ivy-1.1.csv',
        '/content/drive/MyDrive/bug/bug-data/ivy/ivy-1.2.csv',
        '/content/drive/MyDrive/bug/bug-data/log4j/log4j-1.0.csv',
        '/content/drive/MyDrive/bug/bug-data/log4j/log4j-1.1.csv',
        '/content/drive/MyDrive/bug/bug-data/log4j/log4j-1.2.csv',
        '/content/drive/MyDrive/bug/bug-data/lucene/lucene-2.0.csv',
        '/content/drive/MyDrive/bug/bug-data/lucene/lucene-2.2.csv',
        '/content/drive/MyDrive/bug/bug-data/lucene/lucene-2.4.csv',
        '/content/drive/MyDrive/bug/bug-data/poi/poi-1.5.csv',
        '/content/drive/MyDrive/bug/bug-data/poi/poi-2.0.csv',
        '/content/drive/MyDrive/bug/bug-data/poi/poi-2.5.csv',
        '/content/drive/MyDrive/bug/bug-data/poi/poi-3.0.csv',
        '/content/drive/MyDrive/bug/bug-data/synapse/synapse-1.0.csv',
        '/content/drive/MyDrive/bug/bug-data/synapse/synapse-1.1.csv',
        '/content/drive/MyDrive/bug/bug-data/synapse/synapse-1.2.csv',
        '/content/drive/MyDrive/bug/bug-data/velocity/velocity-1.4.csv',
        '/content/drive/MyDrive/bug/bug-data/velocity/velocity-1.5.csv',
        '/content/drive/MyDrive/bug/bug-data/velocity/velocity-1.6.csv',
        '/content/drive/MyDrive/bug/bug-data/xalan/xalan-2.4.csv',
        '/content/drive/MyDrive/bug/bug-data/xalan/xalan-2.5.csv',
        '/content/drive/MyDrive/bug/bug-data/xalan/xalan-2.6.csv',
        '/content/drive/MyDrive/bug/bug-data/xalan/xalan-2.7.csv',
        '/content/drive/MyDrive/bug/bug-data/xerces/xerces-1.1.csv',
        '/content/drive/MyDrive/bug/bug-data/xerces/xerces-1.2.csv',
        '/content/drive/MyDrive/bug/bug-data/xerces/xerces-1.3.csv',
        '/content/drive/MyDrive/bug/bug-data/xerces/xerces-1.4.4.csv',

       ]

    analyzer = BugMetricsAnalyzer(train_file_paths)
    analyzer.analyze_all_metrics()

if __name__ == "__main__":
    main()
````

````python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from typing import List, Dict, Tuple
from collections import defaultdict
from scipy import stats
from adjustText import adjust_text

class BugMetricsAnalyzer:
    def __init__(self, train_paths: List[str]):
        """
        Initialise l'analyseur avec les chemins des fichiers.

        Args:
            train_paths: Liste des chemins vers les fichiers d'entraînement
        """
        self.train_paths = train_paths
        self.metrics = ["loc", "noc", "cbo", "max_cc"]
        self.metric_names = {
            "loc": "Lines of Code",
            "noc": "Number of Classes",
            "cbo": "Coupling Between Objects",
            "max_cc": "Maximum Cyclomatic Complexity"
        }
        self.df_all = None
        self.systems = self._extract_systems()

    def _extract_systems(self) -> Dict[str, List[str]]:
        """
        Extrait les systèmes et leurs versions à partir des chemins de fichiers.

        Returns:
            Dict[str, List[str]]: Dictionnaire des systèmes et leurs versions
        """
        systems_dict = defaultdict(list)
        for path in self.train_paths:
            parts = path.split('/')
            for part in parts:
                if '.csv' in part:
                    system = part.split('-')[0]
                    version = part.split('-')[1].replace('.csv', '')
                    systems_dict[system].append((version, path))

        for system in systems_dict:
            systems_dict[system].sort(key=lambda x: x[0])

        return dict(systems_dict)

    def load_data(self) -> None:
        """Charge et prépare les données pour l'analyse."""
        all_dfs = []
        for system, version_paths in self.systems.items():
            for version, path in version_paths:
                try:
                    df = pd.read_csv(path, usecols=lambda x: x != 'name')
                    df['version'] = version
                    df['system'] = system
                    all_dfs.append(df)
                except Exception as e:
                    print(f"Erreur lors du chargement du fichier {path}: {str(e)}")

        if not all_dfs:
            raise ValueError("Aucune donnée n'a pu être chargée")

        self.df_all = pd.concat(all_dfs, ignore_index=True)

    def calculate_statistics(self, data: pd.Series) -> Dict[str, float]:
        """
        Calcule des statistiques détaillées pour une série de données.

        Args:
            data: Série de données à analyser

        Returns:
            Dict contenant les différentes statistiques calculées
        """
        stats_dict = {
            'variance': data.var(),
            'std_dev': data.std(),
            'mean': data.mean(),
            'median': data.median(),
            'cv': (data.std() / data.mean()) * 100 if data.mean() != 0 else np.nan,
            'q1': data.quantile(0.25),
            'q3': data.quantile(0.75),
            'iqr': data.quantile(0.75) - data.quantile(0.25),
            'skewness': stats.skew(data),
            'kurtosis': stats.kurtosis(data)
        }
        return stats_dict

    def perform_trend_analysis(self, values: np.ndarray) -> Dict[str, float]:
        """
        Effectue une analyse de tendance sur une série de valeurs.

        Args:
            values: Série de valeurs à analyser

        Returns:
            Dict contenant les résultats de l'analyse de tendance
        """
        x = np.arange(len(values))

        try:
            slope, intercept, r_value, p_value, std_err = stats.linregress(x, values)
        except:
            return {
                'slope': np.nan,
                'r_squared': np.nan,
                'p_value': np.nan,
                'trend': 'undefined'
            }

        return {
            'slope': slope,
            'r_squared': r_value ** 2,
            'p_value': p_value,
            'trend': 'increasing' if slope > 0 else 'decreasing' if slope < 0 else 'stable'
        }

    def plot_metric_evolution_by_system(self, metric: str) -> Dict:
        """
        Génère un graphique amélioré et retourne les statistiques pour l'analyse.

        Args:
            metric: Nom de la métrique à analyser

        Returns:
            Dict contenant les tendances et statistiques
        """
        if metric not in self.metrics:
            raise ValueError(f"Métrique invalide: {metric}")

        if self.df_all is None:
            raise ValueError("Les données n'ont pas été chargées. Appelez load_data() d'abord.")

        plt.rcParams['figure.facecolor'] = 'white'
        plt.rcParams['axes.facecolor'] = 'white'
        plt.rcParams['axes.grid'] = True
        plt.rcParams['grid.color'] = '#E6E6E6'
        plt.rcParams['grid.linestyle'] = '--'

        fig = plt.figure(figsize=(20, 12))
        gs = plt.GridSpec(2, 2, height_ratios=[3, 1])
        ax_main = fig.add_subplot(gs[0, :])
        ax_stats = fig.add_subplot(gs[1, :])

        ax_main.grid(True, linestyle='--', alpha=0.7, zorder=0)
        ax_main.set_axisbelow(True)

        colors = plt.cm.tab20(np.linspace(0, 1, len(self.systems)))
        all_stats = []
        system_trends = {}
        texts = []

        for (system, version_paths), color in zip(self.systems.items(), colors):
            system_data = self.df_all[self.df_all['system'] == system]
            grouped_data = system_data.groupby('version')[metric]

            version_stats = []
            x_values = range(len(grouped_data))

            for name, group in grouped_data:
                stats_dict = self.calculate_statistics(group)
                stats_dict['system'] = system
                stats_dict['version'] = name
                version_stats.append(stats_dict)

            if not version_stats:
                print(f"Attention: Pas de données pour le système {system}")
                continue

            variance_values = [stats['variance'] for stats in version_stats]

            line = ax_main.plot(x_values, variance_values,
                              marker='o', label=system, color=color,
                              linewidth=2.5, markersize=8, zorder=2)

            last_x = x_values[-1]
            last_y = variance_values[-1]
            text = ax_main.annotate(system,
                                  (last_x, last_y),
                                  xytext=(10, 0),
                                  textcoords='offset points',
                                  color=color,
                                  fontweight='bold',
                                  bbox=dict(facecolor='white', edgecolor=color, alpha=0.8, pad=2))
            texts.append(text)

            std_dev_values = [stats['std_dev'] for stats in version_stats]
            ax_main.fill_between(x_values,
                               np.array(variance_values) - np.array(std_dev_values),
                               np.array(variance_values) + np.array(std_dev_values),
                               alpha=0.1, color=color, zorder=1)

            trend_analysis = self.perform_trend_analysis(np.array(variance_values))
            system_trends[system] = trend_analysis
            all_stats.extend(version_stats)

        try:
            adjust_text(texts, expand_points=(1.5, 1.5))
        except Exception as e:
            print(f"Attention: Impossible d'ajuster les labels automatiquement: {str(e)}")

        ax_main.set_title(f'Evolution of {self.metric_names[metric]} Variance\nAcross Different Versions',
                         fontsize=14, fontweight='bold', pad=20)
        ax_main.set_xlabel("Version Index", fontsize=12, labelpad=10)
        ax_main.set_ylabel(f"Variance of {self.metric_names[metric]}", fontsize=12, labelpad=10)

        stats_df = pd.DataFrame(all_stats)
        if not stats_df.empty:
            stats_summary = stats_df.select_dtypes(include=[np.number]).agg(['mean', 'min', 'max', 'std']).round(2)

            cell_text = []
            for idx, row in stats_summary.iterrows():
                cell_text.append([f"{val:.2f}" if not pd.isna(val) else "N/A" for val in row])

            table = ax_stats.table(cellText=cell_text,
                                 rowLabels=['Mean', 'Min', 'Max', 'Std'],
                                 colLabels=stats_df.select_dtypes(include=[np.number]).columns,
                                 loc='center',
                                 cellLoc='center')

            table.auto_set_font_size(False)
            table.set_fontsize(9)
            table.scale(1.2, 1.5)

        ax_stats.axis('off')

        trend_text = "Trend Analysis:\n"
        for system, trend in system_trends.items():
            trend_text += (f"{system}: {trend['trend']} (slope: {trend['slope']:.2f}, "
                         f"R²: {trend['r_squared']:.2f}, p-value: {trend['p_value']:.3f})\n")

        plt.figtext(1.02, 0.5, trend_text,
                   bbox=dict(facecolor='white', edgecolor='gray', alpha=0.8),
                   fontsize=10,
                   transform=ax_main.transAxes)

        plt.tight_layout()
        plt.show()

        return {
            'trends': system_trends,
            'statistics': pd.DataFrame(all_stats)
        }

    def analyze_results(self, metric: str, analysis_data: Dict) -> str:
        """
        Analyse et interprète les résultats pour une métrique donnée.

        Args:
            metric: Nom de la métrique analysée
            analysis_data: Dictionnaire contenant les tendances et statistiques

        Returns:
            Rapport d'analyse sous forme de texte
        """
        if not analysis_data['statistics'].empty:
            trends = analysis_data['trends']
            stats_df = analysis_data['statistics']

            analysis = []
            analysis.append(f"\nANALYSE DÉTAILLÉE POUR {self.metric_names[metric].upper()}\n")
            analysis.append("=" * 80 + "\n")

            analysis.append("1. TENDANCES GÉNÉRALES:\n")
            increasing_systems = [sys for sys, trend in trends.items() if trend['trend'] == 'increasing']
            decreasing_systems = [sys for sys, trend in trends.items() if trend['trend'] == 'decreasing']
            stable_systems = [sys for sys, trend in trends.items() if trend['trend'] == 'stable']

            analysis.append(f"- Systèmes avec variance croissante ({len(increasing_systems)}): {', '.join(increasing_systems)}")
            analysis.append(f"- Systèmes avec variance décroissante ({len(decreasing_systems)}): {', '.join(decreasing_systems)}")
            analysis.append(f"- Systèmes stables ({len(stable_systems)}): {', '.join(stable_systems)}\n")

            analysis.append("2. SIGNIFICATIVITÉ DES TENDANCES:\n")
            significant_trends = [(sys, trend) for sys, trend in trends.items() if trend['p_value'] < 0.05]
            if significant_trends:
                analysis.append("Tendances statistiquement significatives (p < 0.05):")
                for sys, trend in significant_trends:
                    analysis.append(f"- {sys}: R² = {trend['r_squared']:.3f}, p-value = {trend['p_value']:.3f}")
            else:
                analysis.append("Aucune tendance n'est statistiquement significative (p < 0.05)")
            analysis.append("")

            analysis.append("3. ANALYSE DE LA DISPERSION:\n")
            if 'cv' in stats_df.columns:
                mean_cv = stats_df['cv'].mean()
                high_cv_systems = stats_df[stats_df['cv'] > mean_cv]['system'].unique()

                analysis.append(f"- Coefficient de variation moyen: {mean_cv:.2f}%")
                analysis.append(f"- Systèmes avec dispersion supérieure à la moyenne: {', '.join(high_cv_systems)}\n")

            analysis.append("4. POINTS D'ATTENTION ET RECOMMANDATIONS:\n")

            concerning_systems = [sys for sys, trend in trends.items()
                                if (trend['trend'] == 'increasing' and trend['p_value'] < 0.05)]

            if concerning_systems:
                analysis.append("Systèmes nécessitant une attention particulière:")
                for sys in concerning_systems:
                    analysis.append(f"- {sys}: Augmentation significative de la variance")

            analysis.append("\nRecommandations:")
            if increasing_systems:
                analysis.append("- Examiner les causes de l'augmentation de la variance dans les systèmes concernés")
            if len(high_cv_systems) > 0:
                analysis.append("- Standardiser les pratiques de développement pour réduire la dispersion")
            if stable_systems:
                analysis.append("- Étudier les bonnes pratiques des systèmes stables pour les appliquer aux autres")

            analysis.append("\n5. INTERPRÉTATION SPÉCIFIQUE:")
            metric_interpretations = {
                'loc': "Une variance élevée dans les lignes de code peut indiquer un manque de standardisation dans la taille des modules.",
                'noc': "La variance dans le nombre de classes peut révéler des incohérences dans la conception orientée objet.",
                'cbo': "Une forte variance dans le couplage suggère des disparités dans l'architecture du système.",
                'max_cc': "La variance de la complexité cyclomatique reflète l'homogénéité des pratiques de codage."
            }
            analysis.append(metric_interpretations.get(metric, "Pas d'interprétation spécifique disponible pour cette métrique."))

            return "\n".join(analysis)
        else:
            return f"Pas de données disponibles pour analyser la métrique {metric}"

    def analyze_all_metrics(self) -> None:
        """Lance l'analyse complète pour toutes les métriques."""
        print("Démarrage de l'analyse des métriques...")
        try:
            self.load_data()

            for metric in self.metrics:
                print(f"\nAnalyse de la métrique : {metric}")
                print("=" * 60)

                try:
                    analysis_data = self.plot_metric_evolution_by_system(metric)
                    interpretation = self.analyze_results(metric, analysis_data)
                    print(interpretation)
                except Exception as e:
                    print(f"Erreur lors de l'analyse de la métrique {metric}: {str(e)}")
                    print("Passage à la métrique suivante...\n")
                    continue

        except Exception as e:
            print(f"Erreur fatale lors de l'analyse: {str(e)}")
            print("Impossible de poursuivre l'analyse.")
            return

        print("\nAnalyse terminée.")

def main():
    # Liste des chemins de fichiers CSV pour l'entrainement
    train_file_paths = [
        '/content/drive/MyDrive/bug/bug-data/jedit/jedit-3.2.csv',
        '/content/drive/MyDrive/bug/bug-data/jedit/jedit-4.0.csv',
        '/content/drive/MyDrive/bug/bug-data/jedit/jedit-4.1.csv',
        '/content/drive/MyDrive/bug/bug-data/jedit/jedit-4.2.csv',
        '/content/drive/MyDrive/bug/bug-data/jedit/jedit-4.3.csv',
        '/content/drive/MyDrive/bug/bug-data/ant/ant-1.3.csv',
        '/content/drive/MyDrive/bug/bug-data/ant/ant-1.4.csv',
        '/content/drive/MyDrive/bug/bug-data/ant/ant-1.5.csv',
        '/content/drive/MyDrive/bug/bug-data/ant/ant-1.6.csv',
        '/content/drive/MyDrive/bug/bug-data/ant/ant-1.7.csv',
        '/content/drive/MyDrive/bug/bug-data/camel/camel-1.0.csv',
        '/content/drive/MyDrive/bug/bug-data/camel/camel-1.2.csv',
        '/content/drive/MyDrive/bug/bug-data/camel/camel-1.4.csv',
        '/content/drive/MyDrive/bug/bug-data/camel/camel-1.6.csv',
        '/content/drive/MyDrive/bug/bug-data/ivy/ivy-1.0.csv',
        '/content/drive/MyDrive/bug/bug-data/ivy/ivy-1.1.csv',
        '/content/drive/MyDrive/bug/bug-data/ivy/ivy-1.2.csv',
        '/content/drive/MyDrive/bug/bug-data/log4j/log4j-1.0.csv',
        '/content/drive/MyDrive/bug/bug-data/log4j/log4j-1.1.csv',
        '/content/drive/MyDrive/bug/bug-data/log4j/log4j-1.2.csv',
        '/content/drive/MyDrive/bug/bug-data/lucene/lucene-2.0.csv',
        '/content/drive/MyDrive/bug/bug-data/lucene/lucene-2.2.csv',
        '/content/drive/MyDrive/bug/bug-data/lucene/lucene-2.4.csv',
        '/content/drive/MyDrive/bug/bug-data/poi/poi-1.5.csv',
        '/content/drive/MyDrive/bug/bug-data/poi/poi-2.0.csv',
        '/content/drive/MyDrive/bug/bug-data/poi/poi-2.5.csv',
        '/content/drive/MyDrive/bug/bug-data/poi/poi-3.0.csv',
        '/content/drive/MyDrive/bug/bug-data/synapse/synapse-1.0.csv',
        '/content/drive/MyDrive/bug/bug-data/synapse/synapse-1.1.csv',
        '/content/drive/MyDrive/bug/bug-data/synapse/synapse-1.2.csv',
        '/content/drive/MyDrive/bug/bug-data/velocity/velocity-1.4.csv',
        '/content/drive/MyDrive/bug/bug-data/velocity/velocity-1.5.csv',
        '/content/drive/MyDrive/bug/bug-data/velocity/velocity-1.6.csv',
        '/content/drive/MyDrive/bug/bug-data/xalan/xalan-2.4.csv',
        '/content/drive/MyDrive/bug/bug-data/xalan/xalan-2.5.csv',
        '/content/drive/MyDrive/bug/bug-data/xalan/xalan-2.6.csv',
        '/content/drive/MyDrive/bug/bug-data/xalan/xalan-2.7.csv',
        '/content/drive/MyDrive/bug/bug-data/xerces/xerces-1.1.csv',
        '/content/drive/MyDrive/bug/bug-data/xerces/xerces-1.2.csv',
        '/content/drive/MyDrive/bug/bug-data/xerces/xerces-1.3.csv',
        '/content/drive/MyDrive/bug/bug-data/xerces/xerces-1.4.4.csv',
    ]

    # Création et exécution de l'analyseur
    analyzer = BugMetricsAnalyzer(train_file_paths)
    analyzer.analyze_all_metrics()

if __name__ == "__main__":
    main()
````

````python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from typing import List, Dict, Tuple
from collections import defaultdict
from scipy import stats
from adjustText import adjust_text

class BugMetricsAnalyzer:
    def __init__(self, train_paths: List[str]):
        """
        Initialise l'analyseur avec les chemins des fichiers.

        Args:
            train_paths: Liste des chemins vers les fichiers d'entraînement
        """
        self.train_paths = train_paths
        self.metrics = ["loc", "noc", "cbo", "max_cc"]
        self.metric_names = {
            "loc": "Lines of Code",
            "noc": "Number of Classes",
            "cbo": "Coupling Between Objects",
            "max_cc": "Maximum Cyclomatic Complexity"
        }
        self.df_all = None
        self.systems = self._extract_systems()

    def _extract_systems(self) -> Dict[str, List[str]]:
        """
        Extrait les systèmes et leurs versions à partir des chemins de fichiers.

        Returns:
            Dict[str, List[str]]: Dictionnaire des systèmes et leurs versions
        """
        systems_dict = defaultdict(list)
        for path in self.train_paths:
            parts = path.split('/')
            for part in parts:
                if '.csv' in part:
                    system = part.split('-')[0]
                    version = part.split('-')[1].replace('.csv', '')
                    systems_dict[system].append((version, path))

        for system in systems_dict:
            systems_dict[system].sort(key=lambda x: x[0])

        return dict(systems_dict)

    def load_data(self) -> None:
        """Charge et prépare les données pour l'analyse."""
        all_dfs = []
        for system, version_paths in self.systems.items():
            for version, path in version_paths:
                try:
                    df = pd.read_csv(path, usecols=lambda x: x != 'name')
                    df['version'] = version
                    df['system'] = system
                    all_dfs.append(df)
                except Exception as e:
                    print(f"Erreur lors du chargement du fichier {path}: {str(e)}")

        if not all_dfs:
            raise ValueError("Aucune donnée n'a pu être chargée")

        self.df_all = pd.concat(all_dfs, ignore_index=True)

    def calculate_statistics(self, data: pd.Series) -> Dict[str, float]:
        """
        Calcule des statistiques détaillées pour une série de données.

        Args:
            data: Série de données à analyser

        Returns:
            Dict contenant les différentes statistiques calculées
        """
        stats_dict = {
            'variance': data.var(),
            'std_dev': data.std(),
            'mean': data.mean(),
            'median': data.median(),
            'cv': (data.std() / data.mean()) * 100 if data.mean() != 0 else np.nan,
            'q1': data.quantile(0.25),
            'q3': data.quantile(0.75),
            'iqr': data.quantile(0.75) - data.quantile(0.25),
            'skewness': stats.skew(data),
            'kurtosis': stats.kurtosis(data)
        }
        return stats_dict

    def perform_trend_analysis(self, values: np.ndarray) -> Dict[str, float]:
        """
        Effectue une analyse de tendance sur une série de valeurs.

        Args:
            values: Série de valeurs à analyser

        Returns:
            Dict contenant les résultats de l'analyse de tendance
        """
        x = np.arange(len(values))

        try:
            slope, intercept, r_value, p_value, std_err = stats.linregress(x, values)
        except:
            return {
                'slope': np.nan,
                'r_squared': np.nan,
                'p_value': np.nan,
                'trend': 'undefined'
            }

        return {
            'slope': slope,
            'r_squared': r_value ** 2,
            'p_value': p_value,
            'trend': 'increasing' if slope > 0 else 'decreasing' if slope < 0 else 'stable'
        }

    def plot_metric_evolution_by_system(self, metric: str) -> Dict:
        """
        Génère un graphique amélioré et retourne les statistiques pour l'analyse.

        Args:
            metric: Nom de la métrique à analyser

        Returns:
            Dict contenant les tendances et statistiques
        """
        if metric not in self.metrics:
            raise ValueError(f"Métrique invalide: {metric}")

        if self.df_all is None:
            raise ValueError("Les données n'ont pas été chargées. Appelez load_data() d'abord.")

        plt.rcParams['figure.facecolor'] = 'white'
        plt.rcParams['axes.facecolor'] = 'white'
        plt.rcParams['axes.grid'] = True
        plt.rcParams['grid.color'] = '#E6E6E6'
        plt.rcParams['grid.linestyle'] = '--'

        fig = plt.figure(figsize=(20, 12))
        gs = plt.GridSpec(2, 2, height_ratios=[3, 1])
        ax_main = fig.add_subplot(gs[0, :])
        ax_stats = fig.add_subplot(gs[1, :])

        ax_main.grid(True, linestyle='--', alpha=0.7, zorder=0)
        ax_main.set_axisbelow(True)

        colors = plt.cm.tab20(np.linspace(0, 1, len(self.systems)))
        all_stats = []
        system_trends = {}
        texts = []

        for (system, version_paths), color in zip(self.systems.items(), colors):
            system_data = self.df_all[self.df_all['system'] == system]
            grouped_data = system_data.groupby('version')[metric]

            version_stats = []
            x_values = range(len(grouped_data))

            for name, group in grouped_data:
                stats_dict = self.calculate_statistics(group)
                stats_dict['system'] = system
                stats_dict['version'] = name
                version_stats.append(stats_dict)

            if not version_stats:
                print(f"Attention: Pas de données pour le système {system}")
                continue

            variance_values = [stats['variance'] for stats in version_stats]

            line = ax_main.plot(x_values, variance_values,
                              marker='o', label=system, color=color,
                              linewidth=2.5, markersize=8, zorder=2)

            last_x = x_values[-1]
            last_y = variance_values[-1]
            text = ax_main.annotate(system,
                                  (last_x, last_y),
                                  xytext=(10, 0),
                                  textcoords='offset points',
                                  color=color,
                                  fontweight='bold',
                                  bbox=dict(facecolor='white', edgecolor=color, alpha=0.8, pad=2))
            texts.append(text)

            std_dev_values = [stats['std_dev'] for stats in version_stats]
            ax_main.fill_between(x_values,
                               np.array(variance_values) - np.array(std_dev_values),
                               np.array(variance_values) + np.array(std_dev_values),
                               alpha=0.1, color=color, zorder=1)

            trend_analysis = self.perform_trend_analysis(np.array(variance_values))
            system_trends[system] = trend_analysis
            all_stats.extend(version_stats)

        try:
            adjust_text(texts, expand_points=(1.5, 1.5))
        except Exception as e:
            print(f"Attention: Impossible d'ajuster les labels automatiquement: {str(e)}")

        ax_main.set_title(f'Evolution of {self.metric_names[metric]} Variance\nAcross Different Versions',
                         fontsize=14, fontweight='bold', pad=20)
        ax_main.set_xlabel("Version Index", fontsize=12, labelpad=10)
        ax_main.set_ylabel(f"Variance of {self.metric_names[metric]}", fontsize=12, labelpad=10)

        stats_df = pd.DataFrame(all_stats)
        if not stats_df.empty:
            stats_summary = stats_df.select_dtypes(include=[np.number]).agg(['mean', 'min', 'max', 'std']).round(2)

            cell_text = []
            for idx, row in stats_summary.iterrows():
                cell_text.append([f"{val:.2f}" if not pd.isna(val) else "N/A" for val in row])

            table = ax_stats.table(cellText=cell_text,
                                 rowLabels=['Mean', 'Min', 'Max', 'Std'],
                                 colLabels=stats_df.select_dtypes(include=[np.number]).columns,
                                 loc='center',
                                 cellLoc='center')

            table.auto_set_font_size(False)
            table.set_fontsize(9)
            table.scale(1.2, 1.5)

        ax_stats.axis('off')

        trend_text = "Trend Analysis:\n"
        for system, trend in system_trends.items():
            trend_text += (f"{system}: {trend['trend']} (slope: {trend['slope']:.2f}, "
                         f"R²: {trend['r_squared']:.2f}, p-value: {trend['p_value']:.3f})\n")

        plt.figtext(1.02, 0.5, trend_text,
                   bbox=dict(facecolor='white', edgecolor='gray', alpha=0.8),
                   fontsize=10,
                   transform=ax_main.transAxes)

        plt.tight_layout()
        plt.show()

        return {
            'trends': system_trends,
            'statistics': pd.DataFrame(all_stats)
        }

    def analyze_results(self, metric: str, analysis_data: Dict) -> str:
        """
        Analyse et interprète les résultats pour une métrique donnée.

        Args:
            metric: Nom de la métrique analysée
            analysis_data: Dictionnaire contenant les tendances et statistiques

        Returns:
            Rapport d'analyse sous forme de texte
        """
        if not analysis_data['statistics'].empty:
            trends = analysis_data['trends']
            stats_df = analysis_data['statistics']

            analysis = []
            analysis.append(f"\nANALYSE DÉTAILLÉE POUR {self.metric_names[metric].upper()}\n")
            analysis.append("=" * 80 + "\n")

            analysis.append("1. TENDANCES GÉNÉRALES:\n")
            increasing_systems = [sys for sys, trend in trends.items() if trend['trend'] == 'increasing']
            decreasing_systems = [sys for sys, trend in trends.items() if trend['trend'] == 'decreasing']
            stable_systems = [sys for sys, trend in trends.items() if trend['trend'] == 'stable']

            analysis.append(f"- Systèmes avec variance croissante ({len(increasing_systems)}): {', '.join(increasing_systems)}")
            analysis.append(f"- Systèmes avec variance décroissante ({len(decreasing_systems)}): {', '.join(decreasing_systems)}")
            analysis.append(f"- Systèmes stables ({len(stable_systems)}): {', '.join(stable_systems)}\n")

            analysis.append("2. SIGNIFICATIVITÉ DES TENDANCES:\n")
            significant_trends = [(sys, trend) for sys, trend in trends.items() if trend['p_value'] < 0.05]
            if significant_trends:
                analysis.append("Tendances statistiquement significatives (p < 0.05):")
                for sys, trend in significant_trends:
                    analysis.append(f"- {sys}: R² = {trend['r_squared']:.3f}, p-value = {trend['p_value']:.3f}")
            else:
                analysis.append("Aucune tendance n'est statistiquement significative (p < 0.05)")
            analysis.append("")

            analysis.append("3. ANALYSE DE LA DISPERSION:\n")
            if 'cv' in stats_df.columns:
                mean_cv = stats_df['cv'].mean()
                high_cv_systems = stats_df[stats_df['cv'] > mean_cv]['system'].unique()

                analysis.append(f"- Coefficient de variation moyen: {mean_cv:.2f}%")
                analysis.append(f"- Systèmes avec dispersion supérieure à la moyenne: {', '.join(high_cv_systems)}\n")

            analysis.append("4. POINTS D'ATTENTION ET RECOMMANDATIONS:\n")

            concerning_systems = [sys for sys, trend in trends.items()
                                if (trend['trend'] == 'increasing' and trend['p_value'] < 0.05)]

            if concerning_systems:
                analysis.append("Systèmes nécessitant une attention particulière:")
                for sys in concerning_systems:
                    analysis.append(f"- {sys}: Augmentation significative de la variance")

            analysis.append("\nRecommandations:")
            if increasing_systems:
                analysis.append("- Examiner les causes de l'augmentation de la variance dans les systèmes concernés")
            if len(high_cv_systems) > 0:
                analysis.append("- Standardiser les pratiques de développement pour réduire la dispersion")
            if stable_systems:
                analysis.append("- Étudier les bonnes pratiques des systèmes stables pour les appliquer aux autres")

            analysis.append("\n5. INTERPRÉTATION SPÉCIFIQUE:")
            metric_interpretations = {
                'loc': "Une variance élevée dans les lignes de code peut indiquer un manque de standardisation dans la taille des modules.",
                'noc': "La variance dans le nombre de classes peut révéler des incohérences dans la conception orientée objet.",
                'cbo': "Une forte variance dans le couplage suggère des disparités dans l'architecture du système.",
                'max_cc': "La variance de la complexité cyclomatique reflète l'homogénéité des pratiques de codage."
            }
            analysis.append(metric_interpretations.get(metric, "Pas d'interprétation spécifique disponible pour cette métrique."))

            return "\n".join(analysis)
        else:
            return f"Pas de données disponibles pour analyser la métrique {metric}"

    def analyze_all_metrics(self) -> None:
        """Lance l'analyse complète pour toutes les métriques."""
        print("Démarrage de l'analyse des métriques...")
        try:
            self.load_data()

            for metric in self.metrics:
                print(f"\nAnalyse de la métrique : {metric}")
                print("=" * 60)

                try:
                    analysis_data = self.plot_metric_evolution_by_system(metric)
                    interpretation = self.analyze_results(metric, analysis_data)
                    print(interpretation)
                except Exception as e:
                    print(f"Erreur lors de l'analyse de la métrique {metric}: {str(e)}")
                    print("Passage à la métrique suivante...\n")
                    continue

        except Exception as e:
            print(f"Erreur fatale lors de l'analyse: {str(e)}")
            print("Impossible de poursuivre l'analyse.")
            return

        print("\nAnalyse terminée.")

def main():
    # Liste des chemins de fichiers CSV pour l'entrainement
    train_file_paths = [
        '/content/drive/MyDrive/bug/bug-data/ant/ant-1.3.csv',
        '/content/drive/MyDrive/bug/bug-data/ant/ant-1.4.csv',
        '/content/drive/MyDrive/bug/bug-data/ant/ant-1.5.csv',
        '/content/drive/MyDrive/bug/bug-data/ant/ant-1.6.csv',
        '/content/drive/MyDrive/bug/bug-data/ant/ant-1.7.csv',
        '/content/drive/MyDrive/bug/bug-data/poi/poi-1.5.csv',
        '/content/drive/MyDrive/bug/bug-data/poi/poi-2.0.csv',
        '/content/drive/MyDrive/bug/bug-data/poi/poi-2.5.csv',
        '/content/drive/MyDrive/bug/bug-data/poi/poi-3.0.csv',
        '/content/drive/MyDrive/bug/bug-data/velocity/velocity-1.4.csv',
        '/content/drive/MyDrive/bug/bug-data/velocity/velocity-1.5.csv',
        '/content/drive/MyDrive/bug/bug-data/velocity/velocity-1.6.csv',
        '/content/drive/MyDrive/bug/bug-data/xalan/xalan-2.4.csv',
        '/content/drive/MyDrive/bug/bug-data/xalan/xalan-2.5.csv',
        '/content/drive/MyDrive/bug/bug-data/xalan/xalan-2.6.csv',
        '/content/drive/MyDrive/bug/bug-data/xalan/xalan-2.7.csv',
    ]

    # Création et exécution de l'analyseur
    analyzer = BugMetricsAnalyzer(train_file_paths)
    analyzer.analyze_all_metrics()

if __name__ == "__main__":
    main()
````

````python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from typing import List, Dict, Tuple
from collections import defaultdict
from scipy import stats

class BugMetricsAnalyzer:
    def __init__(self, train_paths: List[str]):
        self.train_paths = train_paths
        self.metrics = ["loc", "noc", "cbo", "max_cc"]
        self.metric_names = {
            "loc": "Lines of Code",
            "noc": "Number of Classes",
            "cbo": "Coupling Between Objects",
            "max_cc": "Maximum Cyclomatic Complexity"
        }
        self.df_all = None
        self.systems = self._extract_systems()

    def _extract_systems(self) -> Dict[str, List[str]]:
        systems_dict = defaultdict(list)
        for path in self.train_paths:
            parts = path.split('/')
            for part in parts:
                if '.csv' in part:
                    system = part.split('-')[0]
                    version = part.split('-')[1].replace('.csv', '')
                    systems_dict[system].append((version, path))

        for system in systems_dict:
            systems_dict[system].sort(key=lambda x: x[0])

        return dict(systems_dict)

    def load_data(self) -> None:
        all_dfs = []
        for system, version_paths in self.systems.items():
            for version, path in version_paths:
                try:
                    df = pd.read_csv(path, usecols=lambda x: x != 'name')
                    df['version'] = version
                    df['system'] = system
                    all_dfs.append(df)
                except Exception as e:
                    print(f"Erreur lors du chargement du fichier {path}: {str(e)}")

        if not all_dfs:
            raise ValueError("Aucune donnée n'a pu être chargée")

        self.df_all = pd.concat(all_dfs, ignore_index=True)

    def calculate_cv(self, data: pd.Series) -> float:
        """Calcule le coefficient de variation pour une série de données."""
        return (data.std() / data.mean()) * 100 if data.mean() != 0 else np.nan

    def plot_cv_evolution(self) -> Dict:
        """
        Génère un graphique montrant l'évolution des CV pour toutes les métriques.
        """
        if self.df_all is None:
            raise ValueError("Les données n'ont pas été chargées. Appelez load_data() d'abord.")

        # Configuration du style des graphiques
        plt.rcParams['figure.facecolor'] = 'white'
        plt.rcParams['axes.facecolor'] = 'white'
        plt.rcParams['axes.grid'] = True
        plt.rcParams['grid.color'] = '#E6E6E6'
        plt.rcParams['grid.linestyle'] = '--'

        fig, axs = plt.subplots(2, 2, figsize=(20, 16))
        fig.suptitle('Evolution des Coefficients de Variation par Métrique', fontsize=16, y=0.95)

        # Couleurs pour chaque système
        colors = plt.cm.tab10(np.linspace(0, 1, len(self.systems)))

        # Dictionnaire pour stocker les tendances
        all_trends = {}

        # Pour chaque métrique
        for idx, metric in enumerate(self.metrics):
            ax = axs[idx // 2, idx % 2]
            metric_trends = {}

            # Pour chaque système
            for (system, version_paths), color in zip(self.systems.items(), colors):
                system_data = self.df_all[self.df_all['system'] == system]

                # Calcul des CV pour chaque version
                cv_values = []
                versions = []

                for version in sorted(system_data['version'].unique()):
                    version_data = system_data[system_data['version'] == version][metric]
                    cv = self.calculate_cv(version_data)
                    cv_values.append(cv)
                    versions.append(version)

                # Tracer la ligne
                x_values = range(len(cv_values))
                line = ax.plot(x_values, cv_values,
                             marker='o', label=system, color=color,
                             linewidth=2, markersize=6)

            # Ajouter la légende en dehors du graphique
            ax.legend(bbox_to_anchor=(1.05, 1), loc='upper left')

            # Configurer le graphique
            ax.set_title(f'{self.metric_names[metric]}', fontsize=12)
            ax.set_xlabel('Index de Version', fontsize=10)
            ax.set_ylabel('Coefficient de Variation (%)', fontsize=10)
            ax.grid(True, linestyle='--', alpha=0.7)

            # Calculer les tendances pour chaque système dans cette métrique
            for system in self.systems.keys():
                system_data = self.df_all[self.df_all['system'] == system]
                cv_values = []
                for version in sorted(system_data['version'].unique()):
                    version_data = system_data[system_data['version'] == version][metric]
                    cv = self.calculate_cv(version_data)
                    cv_values.append(cv)

                try:
                    x_values = range(len(cv_values))
                    slope, _, r_value, p_value, _ = stats.linregress(x_values, cv_values)
                    trend = {
                        'slope': slope,
                        'r_squared': r_value ** 2,
                        'p_value': p_value,
                        'trend': 'increasing' if slope > 0 else 'decreasing'
                    }
                    metric_trends[system] = trend
                except:
                    metric_trends[system] = {
                        'slope': np.nan,
                        'r_squared': np.nan,
                        'p_value': np.nan,
                        'trend': 'undefined'
                    }

            all_trends[metric] = metric_trends

        # Ajuster la mise en page pour éviter le chevauchement
        plt.tight_layout(rect=[0, 0.03, 0.95, 0.95])
        plt.show()

        return {'trends': all_trends}

    def analyze_cv_results(self, analysis_data: Dict) -> str:
        """Analyse les tendances des coefficients de variation."""
        analysis = []
        analysis.append("\nANALYSE DES COEFFICIENTS DE VARIATION\n")
        analysis.append("=" * 80 + "\n")

        for metric in self.metrics:
            analysis.append(f"\n{self.metric_names[metric].upper()}:")
            trends = analysis_data['trends'][metric]

            # Classifier les systèmes par tendance
            increasing = []
            decreasing = []
            undefined = []
            significant = []

            for system, trend in trends.items():
                if trend['trend'] == 'increasing':
                    increasing.append(system)
                    if trend['p_value'] < 0.05:
                        significant.append((system, trend))
                elif trend['trend'] == 'decreasing':
                    decreasing.append(system)
                    if trend['p_value'] < 0.05:
                        significant.append((system, trend))
                else:
                    undefined.append(system)

            analysis.append(f"\n1. Tendances:")
            analysis.append(f"- Augmentation du CV: {', '.join(increasing) if increasing else 'Aucun'}")
            analysis.append(f"- Diminution du CV: {', '.join(decreasing) if decreasing else 'Aucun'}")
            if undefined:
                analysis.append(f"- Tendance indéfinie: {', '.join(undefined)}")

            analysis.append(f"\n2. Tendances significatives (p < 0.05):")
            if significant:
                for sys, trend in significant:
                    analysis.append(f"- {sys}: pente = {trend['slope']:.3f}, R² = {trend['r_squared']:.3f}")
            else:
                analysis.append("Aucune tendance statistiquement significative")

            analysis.append("\n" + "-" * 40)

        return "\n".join(analysis)

    def analyze_all_cv(self) -> None:
        """Lance l'analyse complète des coefficients de variation."""
        print("Démarrage de l'analyse des coefficients de variation...")
        try:
            self.load_data()
            analysis_data = self.plot_cv_evolution()
            interpretation = self.analyze_cv_results(analysis_data)
            print(interpretation)
        except Exception as e:
            print(f"Erreur lors de l'analyse: {str(e)}")
            return

        print("\nAnalyse terminée.")

def main():
    train_file_paths = [
    '/content/drive/MyDrive/bug/bug-data/jedit/jedit-3.2.csv',
    '/content/drive/MyDrive/bug/bug-data/jedit/jedit-4.0.csv',
    '/content/drive/MyDrive/bug/bug-data/jedit/jedit-4.1.csv',
    '/content/drive/MyDrive/bug/bug-data/jedit/jedit-4.2.csv',
    '/content/drive/MyDrive/bug/bug-data/jedit/jedit-4.3.csv',
    '/content/drive/MyDrive/bug/bug-data/ant/ant-1.3.csv',
    '/content/drive/MyDrive/bug/bug-data/ant/ant-1.4.csv',
    '/content/drive/MyDrive/bug/bug-data/ant/ant-1.5.csv',
    '/content/drive/MyDrive/bug/bug-data/ant/ant-1.6.csv',
    '/content/drive/MyDrive/bug/bug-data/ant/ant-1.7.csv',
    '/content/drive/MyDrive/bug/bug-data/camel/camel-1.0.csv',
    '/content/drive/MyDrive/bug/bug-data/camel/camel-1.2.csv',
    '/content/drive/MyDrive/bug/bug-data/camel/camel-1.4.csv',
    '/content/drive/MyDrive/bug/bug-data/camel/camel-1.6.csv',
    '/content/drive/MyDrive/bug/bug-data/ivy/ivy-1.0.csv',
    '/content/drive/MyDrive/bug/bug-data/ivy/ivy-1.1.csv',
    '/content/drive/MyDrive/bug/bug-data/ivy/ivy-1.2.csv',
    '/content/drive/MyDrive/bug/bug-data/log4j/log4j-1.0.csv',
    '/content/drive/MyDrive/bug/bug-data/log4j/log4j-1.1.csv',
    '/content/drive/MyDrive/bug/bug-data/log4j/log4j-1.2.csv',
    '/content/drive/MyDrive/bug/bug-data/lucene/lucene-2.0.csv',
    '/content/drive/MyDrive/bug/bug-data/lucene/lucene-2.2.csv',
    '/content/drive/MyDrive/bug/bug-data/lucene/lucene-2.4.csv',
    '/content/drive/MyDrive/bug/bug-data/poi/poi-1.5.csv',
    '/content/drive/MyDrive/bug/bug-data/poi/poi-2.0.csv',
    '/content/drive/MyDrive/bug/bug-data/poi/poi-2.5.csv',
    '/content/drive/MyDrive/bug/bug-data/poi/poi-3.0.csv',
    '/content/drive/MyDrive/bug/bug-data/synapse/synapse-1.0.csv',
    '/content/drive/MyDrive/bug/bug-data/synapse/synapse-1.1.csv',
    '/content/drive/MyDrive/bug/bug-data/synapse/synapse-1.2.csv',
    '/content/drive/MyDrive/bug/bug-data/velocity/velocity-1.4.csv',
    '/content/drive/MyDrive/bug/bug-data/velocity/velocity-1.5.csv',
    '/content/drive/MyDrive/bug/bug-data/velocity/velocity-1.6.csv',
    '/content/drive/MyDrive/bug/bug-data/xalan/xalan-2.4.csv',
    '/content/drive/MyDrive/bug/bug-data/xalan/xalan-2.5.csv',
    '/content/drive/MyDrive/bug/bug-data/xalan/xalan-2.6.csv',
    '/content/drive/MyDrive/bug/bug-data/xerces/xerces-1.1.csv',
    '/content/drive/MyDrive/bug/bug-data/xerces/xerces-1.2.csv',
    '/content/drive/MyDrive/bug/bug-data/xerces/xerces-1.3.csv',
    '/content/drive/MyDrive/bug/bug-data/xerces/xerces-1.4.4.csv',
    ]

    analyzer = BugMetricsAnalyzer(train_file_paths)
    analyzer.analyze_all_cv()

if __name__ == "__main__":
    main()
````

````python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from typing import List, Dict, Tuple
from collections import defaultdict
from scipy import stats

class BugMetricsAnalyzer:
    def __init__(self, train_paths: List[str]):
        self.train_paths = train_paths
        self.metrics = ["loc", "noc", "cbo", "max_cc"]
        self.metric_names = {
            "loc": "Lines of Code",
            "noc": "Number of Classes",
            "cbo": "Coupling Between Objects",
            "max_cc": "Maximum Cyclomatic Complexity"
        }
        self.df_all = None
        self.systems = self._extract_systems()

    def _extract_systems(self) -> Dict[str, List[str]]:
        systems_dict = defaultdict(list)
        for path in self.train_paths:
            parts = path.split('/')
            for part in parts:
                if '.csv' in part:
                    system = part.split('-')[0]
                    version = part.split('-')[1].replace('.csv', '')
                    systems_dict[system].append((version, path))

        for system in systems_dict:
            systems_dict[system].sort(key=lambda x: x[0])

        return dict(systems_dict)

    def load_data(self) -> None:
        all_dfs = []
        for system, version_paths in self.systems.items():
            for version, path in version_paths:
                try:
                    df = pd.read_csv(path, usecols=lambda x: x != 'name')
                    df['version'] = version
                    df['system'] = system
                    all_dfs.append(df)
                except Exception as e:
                    print(f"Erreur lors du chargement du fichier {path}: {str(e)}")

        if not all_dfs:
            raise ValueError("Aucune donnée n'a pu être chargée")

        self.df_all = pd.concat(all_dfs, ignore_index=True)

    def calculate_cv(self, data: pd.Series) -> float:
        """Calcule le coefficient de variation pour une série de données."""
        return (data.std() / data.mean()) * 100 if data.mean() != 0 else np.nan

    def plot_cv_evolution(self) -> Dict:
        """
        Génère un graphique montrant l'évolution des CV pour toutes les métriques.
        """
        if self.df_all is None:
            raise ValueError("Les données n'ont pas été chargées. Appelez load_data() d'abord.")

        # Configuration du style des graphiques
        plt.rcParams['figure.facecolor'] = 'white'
        plt.rcParams['axes.facecolor'] = 'white'
        plt.rcParams['axes.grid'] = True
        plt.rcParams['grid.color'] = '#E6E6E6'
        plt.rcParams['grid.linestyle'] = '--'

        fig, axs = plt.subplots(2, 2, figsize=(20, 16))
        fig.suptitle('Evolution des Coefficients de Variation par Métrique', fontsize=16, y=0.95)

        # Couleurs pour chaque système
        colors = plt.cm.tab10(np.linspace(0, 1, len(self.systems)))

        # Dictionnaire pour stocker les tendances
        all_trends = {}

        # Pour chaque métrique
        for idx, metric in enumerate(self.metrics):
            ax = axs[idx // 2, idx % 2]
            metric_trends = {}

            # Pour chaque système
            for (system, version_paths), color in zip(self.systems.items(), colors):
                system_data = self.df_all[self.df_all['system'] == system]

                # Calcul des CV pour chaque version
                cv_values = []
                versions = []

                for version in sorted(system_data['version'].unique()):
                    version_data = system_data[system_data['version'] == version][metric]
                    cv = self.calculate_cv(version_data)
                    cv_values.append(cv)
                    versions.append(version)

                # Tracer la ligne
                x_values = range(len(cv_values))
                line = ax.plot(x_values, cv_values,
                             marker='o', label=system, color=color,
                             linewidth=2, markersize=6)

            # Ajouter la légende en dehors du graphique
            ax.legend(bbox_to_anchor=(1.05, 1), loc='upper left')

            # Configurer le graphique
            ax.set_title(f'{self.metric_names[metric]}', fontsize=12)
            ax.set_xlabel('Index de Version', fontsize=10)
            ax.set_ylabel('Coefficient de Variation (%)', fontsize=10)
            ax.grid(True, linestyle='--', alpha=0.7)

            # Calculer les tendances pour chaque système dans cette métrique
            for system in self.systems.keys():
                system_data = self.df_all[self.df_all['system'] == system]
                cv_values = []
                for version in sorted(system_data['version'].unique()):
                    version_data = system_data[system_data['version'] == version][metric]
                    cv = self.calculate_cv(version_data)
                    cv_values.append(cv)

                try:
                    x_values = range(len(cv_values))
                    slope, _, r_value, p_value, _ = stats.linregress(x_values, cv_values)
                    trend = {
                        'slope': slope,
                        'r_squared': r_value ** 2,
                        'p_value': p_value,
                        'trend': 'increasing' if slope > 0 else 'decreasing'
                    }
                    metric_trends[system] = trend
                except:
                    metric_trends[system] = {
                        'slope': np.nan,
                        'r_squared': np.nan,
                        'p_value': np.nan,
                        'trend': 'undefined'
                    }

            all_trends[metric] = metric_trends

        # Ajuster la mise en page pour éviter le chevauchement
        plt.tight_layout(rect=[0, 0.03, 0.95, 0.95])
        plt.show()

        return {'trends': all_trends}

    def analyze_cv_results(self, analysis_data: Dict) -> str:
        """Analyse les tendances des coefficients de variation."""
        analysis = []
        analysis.append("\nANALYSE DES COEFFICIENTS DE VARIATION\n")
        analysis.append("=" * 80 + "\n")

        for metric in self.metrics:
            analysis.append(f"\n{self.metric_names[metric].upper()}:")
            trends = analysis_data['trends'][metric]

            # Classifier les systèmes par tendance
            increasing = []
            decreasing = []
            undefined = []
            significant = []

            for system, trend in trends.items():
                if trend['trend'] == 'increasing':
                    increasing.append(system)
                    if trend['p_value'] < 0.05:
                        significant.append((system, trend))
                elif trend['trend'] == 'decreasing':
                    decreasing.append(system)
                    if trend['p_value'] < 0.05:
                        significant.append((system, trend))
                else:
                    undefined.append(system)

            analysis.append(f"\n1. Tendances:")
            analysis.append(f"- Augmentation du CV: {', '.join(increasing) if increasing else 'Aucun'}")
            analysis.append(f"- Diminution du CV: {', '.join(decreasing) if decreasing else 'Aucun'}")
            if undefined:
                analysis.append(f"- Tendance indéfinie: {', '.join(undefined)}")

            analysis.append(f"\n2. Tendances significatives (p < 0.05):")
            if significant:
                for sys, trend in significant:
                    analysis.append(f"- {sys}: pente = {trend['slope']:.3f}, R² = {trend['r_squared']:.3f}")
            else:
                analysis.append("Aucune tendance statistiquement significative")

            analysis.append("\n" + "-" * 40)

        return "\n".join(analysis)

    def analyze_all_cv(self) -> None:
        """Lance l'analyse complète des coefficients de variation."""
        print("Démarrage de l'analyse des coefficients de variation...")
        try:
            self.load_data()
            analysis_data = self.plot_cv_evolution()
            interpretation = self.analyze_cv_results(analysis_data)
            print(interpretation)
        except Exception as e:
            print(f"Erreur lors de l'analyse: {str(e)}")
            return

        print("\nAnalyse terminée.")

def main():
    train_file_paths = [
        '/content/drive/MyDrive/bug/bug-data/ant/ant-1.3.csv',
        '/content/drive/MyDrive/bug/bug-data/ant/ant-1.4.csv',
        '/content/drive/MyDrive/bug/bug-data/ant/ant-1.5.csv',
        '/content/drive/MyDrive/bug/bug-data/ant/ant-1.6.csv',
        '/content/drive/MyDrive/bug/bug-data/ant/ant-1.7.csv',
        '/content/drive/MyDrive/bug/bug-data/poi/poi-1.5.csv',
        '/content/drive/MyDrive/bug/bug-data/poi/poi-2.0.csv',
        '/content/drive/MyDrive/bug/bug-data/poi/poi-2.5.csv',
        '/content/drive/MyDrive/bug/bug-data/poi/poi-3.0.csv',
        '/content/drive/MyDrive/bug/bug-data/velocity/velocity-1.4.csv',
        '/content/drive/MyDrive/bug/bug-data/velocity/velocity-1.5.csv',
        '/content/drive/MyDrive/bug/bug-data/velocity/velocity-1.6.csv',
        '/content/drive/MyDrive/bug/bug-data/xalan/xalan-2.4.csv',
        '/content/drive/MyDrive/bug/bug-data/xalan/xalan-2.5.csv',
        '/content/drive/MyDrive/bug/bug-data/xalan/xalan-2.6.csv',
        '/content/drive/MyDrive/bug/bug-data/xalan/xalan-2.7.csv',
    ]

    analyzer = BugMetricsAnalyzer(train_file_paths)
    analyzer.analyze_all_cv()

if __name__ == "__main__":
    main()
````

````python
import pandas as pd
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
import seaborn as sns
from statsmodels.tsa.stattools import adfuller
from statsmodels.stats.diagnostic import acorr_ljungbox
from statsmodels.tsa.seasonal import seasonal_decompose
from scipy.stats import kruskal, friedmanchisquare
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
from scipy.cluster.hierarchy import dendrogram, linkage

class AdvancedMetricsAnalyzer:
    def __init__(self, df_all):
        """
        Initialise l'analyseur de métriques

        Parameters:
        df_all (pd.DataFrame): DataFrame contenant les métriques
        """
        if not isinstance(df_all, pd.DataFrame):
            raise TypeError("Les données doivent être un DataFrame pandas")

        required_columns = ["system", "loc", "noc", "cbo", "max_cc"]
        missing_columns = [col for col in required_columns if col not in df_all.columns]
        if missing_columns:
            raise ValueError(f"Colonnes manquantes: {missing_columns}")

        # Supprimer les lignes avec des valeurs manquantes
        self.df_all = df_all.dropna(subset=["loc", "noc", "cbo", "max_cc"])
        self.metrics = ["loc", "noc", "cbo", "max_cc"]

        # Vérifier qu'il reste des données après nettoyage
        if len(self.df_all) == 0:
            raise ValueError("Aucune donnée valide après nettoyage")

    def _validate_data_for_analysis(self, data, min_samples=3):
        """Valide les données pour l'analyse statistique"""
        if isinstance(data, pd.DataFrame):
            if len(data) < min_samples:
                raise ValueError(f"Pas assez de données pour l'analyse (minimum {min_samples} requis)")
            if data.isnull().values.any():
                raise ValueError("Présence de valeurs manquantes dans les données")
        elif isinstance(data, pd.Series):
            if len(data) < min_samples:
                raise ValueError(f"Pas assez de données pour l'analyse (minimum {min_samples} requis)")
            if data.isnull().any():
                raise ValueError("Présence de valeurs manquantes dans les données")
        else:
            raise TypeError("Les données doivent être un DataFrame ou une Series pandas")

    def test_distributions(self, system=None):
        """Analyse la distribution des métriques"""
        try:
            results = {}

            # Filtrer les données si un système est spécifié
            if system is not None:
                if system not in self.df_all['system'].unique():
                    raise ValueError(f"Système '{system}' non trouvé dans les données")
                data = self.df_all[self.df_all['system'] == system].copy()
            else:
                data = self.df_all.copy()

            # Vérifier qu'il y a des données à analyser
            if len(data) == 0:
                raise ValueError("Aucune donnée à analyser")

            # Analyser chaque métrique
            for metric in self.metrics:
                metric_data = data[metric]
                self._validate_data_for_analysis(metric_data)

                # Tests de normalité
                shapiro_stat, shapiro_p = stats.shapiro(metric_data.values)
                ks_stat, ks_p = stats.kstest(metric_data.values, 'norm')

                # Statistiques descriptives
                skewness = stats.skew(metric_data.values)
                kurtosis = stats.kurtosis(metric_data.values)

                results[metric] = {
                    'shapiro_test': {'statistic': shapiro_stat, 'p_value': shapiro_p},
                    'ks_test': {'statistic': ks_stat, 'p_value': ks_p},
                    'skewness': skewness,
                    'kurtosis': kurtosis,
                    'basic_stats': {
                        'mean': metric_data.mean(),
                        'median': metric_data.median(),
                        'std': metric_data.std(),
                        'min': metric_data.min(),
                        'max': metric_data.max()
                    }
                }

                # QQ Plot
                plt.figure(figsize=(10, 6))
                stats.probplot(metric_data.values, dist="norm", plot=plt)
                plt.title(f'Q-Q Plot for {metric}')
                plt.show()

            return results

        except Exception as e:
            print(f"Erreur dans l'analyse des distributions : {str(e)}")
            raise

    def _interpret_correlation(self, corr, p_value, alpha=0.05):
        """Interprète les résultats de corrélation"""
        strength = ""
        if abs(corr) < 0.1:
            strength = "négligeable"
        elif abs(corr) < 0.3:
            strength = "faible"
        elif abs(corr) < 0.5:
            strength = "modérée"
        elif abs(corr) < 0.7:
            strength = "forte"
        else:
            strength = "très forte"

        significance = "statistiquement significative" if p_value < alpha else "non significative"
        direction = "positive" if corr > 0 else "négative"

        return f"Corrélation {direction} {strength} ({significance})"

    def test_distributions(self, system=None):
        """Analyse la distribution des métriques"""
        results = {}

        data = self.df_all if system is None else self.df_all[self.df_all['system'] == system]
        if len(data) == 0:
            raise ValueError(f"Aucune donnée trouvée pour le système {system}")

        for metric in self.metrics:
            metric_data = data[metric].values
            self._validate_data_for_analysis(pd.Series(metric_data))

            # Tests de normalité
            shapiro_stat, shapiro_p = stats.shapiro(metric_data)
            ks_stat, ks_p = stats.kstest(metric_data, 'norm')

            # Statistiques descriptives
            skewness = stats.skew(metric_data)
            kurtosis = stats.kurtosis(metric_data)

            results[metric] = {
                'shapiro_test': {'statistic': shapiro_stat, 'p_value': shapiro_p},
                'ks_test': {'statistic': ks_stat, 'p_value': ks_p},
                'skewness': skewness,
                'kurtosis': kurtosis,
                'basic_stats': {
                    'mean': np.mean(metric_data),
                    'median': np.median(metric_data),
                    'std': np.std(metric_data),
                    'min': np.min(metric_data),
                    'max': np.max(metric_data)
                }
            }

            # QQ Plot
            plt.figure(figsize=(10, 6))
            stats.probplot(metric_data, dist="norm", plot=plt)
            plt.title(f'Q-Q Plot for {metric}')
            plt.show()

        return results

def correlation_analysis(self):

    """Analyse des corrélations entre métriques"""
    # Vérifier les données
    self._validate_data_for_analysis(self.df_all[self.metrics])

    # Matrice de corrélation avec différentes méthodes
    correlation_methods = ['pearson', 'spearman']
    correlation_results = {}

    for method in correlation_methods:
        # Calcul de la matrice de corrélation
        corr_matrix = self.df_all[self.metrics].corr(method=method)

        # Visualisation
        plt.figure(figsize=(10, 8))
        sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', vmin=-1, vmax=1, fmt='.3f')
        plt.title(f'Correlation Matrix ({method.capitalize()})')
        plt.show()

        # Tests de corrélation
        tests = {}
        # Itérer sur les paires de métriques uniques
        for i, m1 in enumerate(self.metrics):
            for j, m2 in enumerate(self.metrics[i+1:], i+1):  # évite les doublons et comparaisons avec soi-même
                if method == 'pearson':
                    corr, p_value = stats.pearsonr(self.df_all[m1].values, self.df_all[m2].values)
                else:
                    corr, p_value = stats.spearmanr(self.df_all[m1].values, self.df_all[m2].values)

                tests[f'{m1}_vs_{m2}'] = {
                    'correlation': corr,
                    'p_value': p_value,
                    'interpretation': self._interpret_correlation(corr, p_value)
                }

        correlation_results[method] = tests

    return correlation_results

    def _interpret_correlation(self, corr, p_value, alpha=0.05):
        """Interprète les résultats de corrélation"""
        strength = ""
        if abs(corr) < 0.1:
            strength = "négligeable"
        elif abs(corr) < 0.3:
            strength = "faible"
        elif abs(corr) < 0.5:
            strength = "modérée"
        elif abs(corr) < 0.7:
            strength = "forte"
        else:
            strength = "très forte"

        significance = "statistiquement significative" if p_value < alpha else "non significative"
        direction = "positive" if corr > 0 else "négative"

        return f"Corrélation {direction} {strength} ({significance})"

    def trend_analysis(self, system):
        """Analyse approfondie des tendances pour un système"""
        if system not in self.df_all['system'].unique():
            raise ValueError(f"Système '{system}' non trouvé dans les données")

        system_data = self.df_all[self.df_all['system'] == system]
        self._validate_data_for_analysis(system_data[self.metrics])

        results = {}
        for metric in self.metrics:
            metric_data = system_data[metric].values

            try:
                # Test de Mann-Kendall
                mk_result = stats.kendalltau(range(len(metric_data)), metric_data)

                # Test de stationnarité
                adf_result = adfuller(metric_data)

                # Test d'autocorrélation
                lb_result = acorr_ljungbox(metric_data)

                results[metric] = {
                    'mann_kendall': {
                        'tau': mk_result[0],
                        'p_value': mk_result[1],
                        'interpretation': 'Tendance significative' if mk_result[1] < 0.05 else 'Pas de tendance significative'
                    },
                    'adf_test': {
                        'statistic': adf_result[0],
                        'p_value': adf_result[1],
                        'interpretation': 'Série stationnaire' if adf_result[1] < 0.05 else 'Série non stationnaire'
                    },
                    'ljung_box': {
                        'statistic': lb_result.statistic[0],
                        'p_value': lb_result.pvalue[0],
                        'interpretation': 'Présence d\'autocorrélation' if lb_result.pvalue[0] < 0.05 else 'Pas d\'autocorrélation'
                    }
                }

            except Exception as e:
                results[metric] = {'error': str(e)}

        return results

    def compare_systems(self):
        """Compare les systèmes entre eux"""
        if 'system' not in self.df_all.columns:
            raise ValueError("Colonne 'system' manquante")

        systems = self.df_all['system'].unique()
        if len(systems) < 2:
            raise ValueError("Au moins deux systèmes sont nécessaires pour la comparaison")

        results = {}
        for metric in self.metrics:
            try:
                # Préparation des données pour Kruskal-Wallis
                systems_data = [group[metric].values for name, group in self.df_all.groupby('system')]

                # Test de Kruskal-Wallis
                h_stat, p_value = kruskal(*systems_data)

                results[metric] = {
                    'kruskal_test': {
                        'statistic': h_stat,
                        'p_value': p_value,
                        'interpretation': 'Différences significatives entre les systèmes' if p_value < 0.05 else 'Pas de différences significatives'
                    }
                }

                # Visualisation par boîtes à moustaches
                plt.figure(figsize=(12, 6))
                sns.boxplot(x='system', y=metric, data=self.df_all)
                plt.title(f'Distribution of {metric} across systems')
                plt.xticks(rotation=45)
                plt.show()

            except Exception as e:
                results[metric] = {'error': str(e)}

        return results

    def multivariate_analysis(self):
        """Analyse multivariée des métriques"""
        try:
            # Vérification des données
            metrics_df = self.df_all[self.metrics]
            self._validate_data_for_analysis(metrics_df)

            # Standardisation
            scaler = StandardScaler()
            scaled_data = scaler.fit_transform(metrics_df)

            # ACP
            pca = PCA()
            pca_result = pca.fit_transform(scaled_data)

            # Calcul des résultats
            explained_variance = pca.explained_variance_ratio_
            cumulative_variance = np.cumsum(explained_variance)

            # Visualisation ACP
            plt.figure(figsize=(10, 6))
            plt.plot(range(1, len(explained_variance) + 1), cumulative_variance, 'bo-')
            plt.xlabel('Nombre de composantes')
            plt.ylabel('Variance expliquée cumulée')
            plt.title('Analyse en Composantes Principales')
            plt.grid(True)
            plt.show()

            # Clustering hiérarchique
            linkage_matrix = linkage(scaled_data, method='ward')

            plt.figure(figsize=(12, 8))
            dendrogram(linkage_matrix)
            plt.title('Dendrogramme du clustering hiérarchique')
            plt.xlabel('Index d\'échantillon')
            plt.ylabel('Distance')
            plt.show()

            # Préparer les résultats détaillés
            results = {
                'explained_variance': explained_variance.tolist(),
                'cumulative_variance': cumulative_variance.tolist(),
                'components': pca.components_.tolist(),
                'feature_importance': {
                    metric: abs(pca.components_[0][i])
                    for i, metric in enumerate(self.metrics)
                }
            }

            return results

        except Exception as e:
            return {'error': str(e)}

def read_and_prepare_data(file_paths):
    """Lit et prépare les données à partir des fichiers CSV"""
    all_data = []

    for file_path in file_paths:
        try:
            # Extraire le système et la version du chemin
            parts = file_path.split('/')
            system = parts[-2]
            version = parts[-1].split('.csv')[0]

            # Lire le fichier CSV
            df = pd.read_csv(file_path)

            # Ajouter les informations système et version
            df['system'] = system
            df['version'] = version

            all_data.append(df)

        except Exception as e:
            print(f"Erreur lors de la lecture de {file_path}: {str(e)}")
            continue

    if not all_data:
        raise ValueError("Aucun fichier n'a pu être lu correctement")

    return pd.concat(all_data, ignore_index=True)

def main():
    # Liste des chemins de fichiers
    train_file_paths = [
    '/content/drive/MyDrive/bug/bug-data/jedit/jedit-3.2.csv',
    '/content/drive/MyDrive/bug/bug-data/jedit/jedit-4.0.csv',
    '/content/drive/MyDrive/bug/bug-data/jedit/jedit-4.1.csv',
    '/content/drive/MyDrive/bug/bug-data/jedit/jedit-4.2.csv',
    '/content/drive/MyDrive/bug/bug-data/jedit/jedit-4.3.csv',
    '/content/drive/MyDrive/bug/bug-data/ant/ant-1.3.csv',
    '/content/drive/MyDrive/bug/bug-data/ant/ant-1.4.csv',
    '/content/drive/MyDrive/bug/bug-data/ant/ant-1.5.csv',
    '/content/drive/MyDrive/bug/bug-data/ant/ant-1.6.csv',
    '/content/drive/MyDrive/bug/bug-data/ant/ant-1.7.csv',
    '/content/drive/MyDrive/bug/bug-data/camel/camel-1.0.csv',
    '/content/drive/MyDrive/bug/bug-data/camel/camel-1.2.csv',
    '/content/drive/MyDrive/bug/bug-data/camel/camel-1.4.csv',
    '/content/drive/MyDrive/bug/bug-data/camel/camel-1.6.csv',
    '/content/drive/MyDrive/bug/bug-data/ivy/ivy-1.0.csv',
    '/content/drive/MyDrive/bug/bug-data/ivy/ivy-1.1.csv',
    '/content/drive/MyDrive/bug/bug-data/ivy/ivy-1.2.csv',
    '/content/drive/MyDrive/bug/bug-data/log4j/log4j-1.0.csv',
    '/content/drive/MyDrive/bug/bug-data/log4j/log4j-1.1.csv',
    '/content/drive/MyDrive/bug/bug-data/log4j/log4j-1.2.csv',
    '/content/drive/MyDrive/bug/bug-data/lucene/lucene-2.0.csv',
    '/content/drive/MyDrive/bug/bug-data/lucene/lucene-2.2.csv',
    '/content/drive/MyDrive/bug/bug-data/lucene/lucene-2.4.csv',
    '/content/drive/MyDrive/bug/bug-data/poi/poi-1.5.csv',
    '/content/drive/MyDrive/bug/bug-data/poi/poi-2.0.csv',
    '/content/drive/MyDrive/bug/bug-data/poi/poi-2.5.csv',
    '/content/drive/MyDrive/bug/bug-data/poi/poi-3.0.csv',
    '/content/drive/MyDrive/bug/bug-data/synapse/synapse-1.0.csv',
    '/content/drive/MyDrive/bug/bug-data/synapse/synapse-1.1.csv',
    '/content/drive/MyDrive/bug/bug-data/synapse/synapse-1.2.csv',
    '/content/drive/MyDrive/bug/bug-data/velocity/velocity-1.4.csv',
    '/content/drive/MyDrive/bug/bug-data/velocity/velocity-1.5.csv',
    '/content/drive/MyDrive/bug/bug-data/velocity/velocity-1.6.csv',
    '/content/drive/MyDrive/bug/bug-data/xalan/xalan-2.4.csv',
    '/content/drive/MyDrive/bug/bug-data/xalan/xalan-2.5.csv',
    '/content/drive/MyDrive/bug/bug-data/xalan/xalan-2.6.csv',
    '/content/drive/MyDrive/bug/bug-data/xerces/xerces-1.1.csv',
    '/content/drive/MyDrive/bug/bug-data/xerces/xerces-1.2.csv',
    '/content/drive/MyDrive/bug/bug-data/xerces/xerces-1.3.csv',
    '/content/drive/MyDrive/bug/bug-data/xerces/xerces-1.4.4.csv',
    ]

    try:
        # Lire et préparer les données
        combined_df = read_and_prepare_data(train_file_paths)
        print(f"Données chargées : {len(combined_df)} entrées")

        # Créer l'analyseur
        analyzer = AdvancedMetricsAnalyzer(combined_df)


        # Exécuter l'analyse des distributions
        results = analyzer.test_distributions()

        # Exécuter l'analyse des corrélations
        correlation_results = analyzer.correlation_analysis()

        # Effectuer les analyses
        results = {}

        print("\nAnalyse des distributions...")
        results['distributions'] = analyzer.test_distributions()

        print("\nAnalyse des corrélations...")
        results['correlations'] = analyzer.correlation_analysis()

        print("\nAnalyse des tendances...")
        results['trends'] = analyzer.trend_analysis('ant')

        print("\nComparaison des systèmes...")
        results['comparisons'] = analyzer.compare_systems()

        print("\nAnalyse multivariée...")
        results['multivariate'] = analyzer.multivariate_analysis()

        # Sauvegarder les résultats
        with open('analysis_results.txt', 'w') as f:
            for analysis_type, analysis_results in results.items():
                f.write(f"\n{'='*50}\n")
                f.write(f"{analysis_type.upper()}\n")
                f.write(f"{'='*50}\n")
                f.write(str(analysis_results))

    except Exception as e:
        print(f"Une erreur s'est produite : {str(e)}")

if __name__ == "__main__":
    main()
````
